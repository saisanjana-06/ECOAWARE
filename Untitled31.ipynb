```python

# ============================================
# SYNDATA ARENA - CELL 2
# PYTHON AI ENGINE + DATA ANALYSIS
# NO GEMINI API DEPENDENCY
# ============================================

import os
import io
import json
import time
import warnings
import traceback
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import gradio as gr

from PIL import Image

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
    accuracy_score,
    classification_report
)

warnings.filterwarnings("ignore")

# ============================================
# AI ENGINE CONFIGURATION
# ============================================

AI_ENGINE = "Python Intelligent Analysis Engine"
GEMINI_READY = False
client = None
GEMINI_API_KEY = None

print("Gemini dependency: DISABLED")
print("AI Engine:", AI_ENGINE)


# ============================================
# AUTOMATIC DATASET INSIGHTS
# ============================================

def generate_python_insights(df):

    if df is None or df.empty:
        return "No dataset available for analysis."

    insights = []

    rows, cols = df.shape

    insights.append("## AI-Generated Dataset Insights")
    insights.append(
        f"The dataset contains **{rows:,} records** "
        f"and **{cols} columns**."
    )

    numerical = df.select_dtypes(include=np.number)
    categorical = df.select_dtypes(
        include=["object", "category", "bool"]
    )

    insights.append(
        f"- Numerical columns: {len(numerical)}"
    )
    insights.append(
        f"- Categorical columns: {len(categorical)}"
    )

    # Missing values
    missing = df.isnull().sum()
    total_missing = int(missing.sum())

    if total_missing:
        insights.append(
            f"- **Missing values:** {total_missing} cells "
            "need attention."
        )
        for col, count in missing[missing > 0].items():
            insights.append(f"  - `{col}`: {count} missing values.")
    else:
        insights.append("- **Missing values:** None detected.")

    # Duplicate records
    duplicates = int(df.duplicated().sum())

    insights.append(
        f"- **Duplicate records:** {duplicates}"
    )

    if duplicates:
        insights.append(
            "Consider removing duplicate records before modelling."
        )

    # Numerical analysis
    if not numerical.empty:

        insights.append("\n### Numerical Column Insights")

        for col in numerical.columns:

            series = numerical[col].dropna()

            if series.empty:
                continue

            mean_value = series.mean()
            min_value = series.min()
            max_value = series.max()
            std_value = series.std()

            insights.append(
                f"- **{col}:** Average = {mean_value:.2f}, "
                f"Minimum = {min_value:.2f}, "
                f"Maximum = {max_value:.2f}."
            )

            if std_value == 0 or pd.isna(std_value):
                insights.append(
                    f"  - `{col}` has constant values."
                )

            elif std_value > abs(mean_value) * 2:
                insights.append(
                    f"  - `{col}` has relatively high variation."
                )

    # Categorical analysis
    if not categorical.empty:

        insights.append("\n### Categorical Column Insights")

        for col in categorical.columns:

            counts = df[col].value_counts(dropna=True)

            if counts.empty:
                continue

            top_value = str(counts.index[0])

            insights.append(
                f"- **{col}:** {counts.nunique()} distinct "
                f"value(s). Most frequent value: "
                f"`{top_value}` ({counts.iloc[0]} records)."
            )

    # Correlation analysis
    if len(numerical.columns) >= 2:

        corr = numerical.corr()

        pairs = corr.where(
            np.triu(np.ones(corr.shape), k=1).astype(bool)
        ).stack()

        if not pairs.empty:

            strongest_pair = pairs.abs().idxmax()
            correlation = pairs.loc[strongest_pair]

            insights.append("\n### Correlation Insight")

            insights.append(
                f"`{strongest_pair[0]}` and "
                f"`{strongest_pair[1]}` have a correlation "
                f"of **{correlation:.2f}**."
            )

            if abs(correlation) >= 0.7:
                insights.append(
                    "This indicates a strong relationship "
                    "between these two numerical columns."
                )

    insights.append(
        "\n### Recommendation\n"
        "Review missing values, duplicate records and unusual "
        "variations before proceeding with machine learning."
    )

    return "\n".join(insights)


# ============================================
# LIGHTWEIGHT DATASET-AWARE CHATBOT
# ============================================

def python_synbot_response(message, df=None):

    if not message or not str(message).strip():
        return "Please enter a question."

    question = str(message).lower().strip()

    if df is None or df.empty:
        return (
            "Hello! I am SynBot. Upload a CSV dataset or image "
            "to start exploring your data."
        )

    rows, cols = df.shape

    numerical = df.select_dtypes(include=np.number)
    categorical = df.select_dtypes(
        include=["object", "category", "bool"]
    )

    if any(word in question for word in ["hello", "hi", "hey"]):
        return (
            "Hello! I am SynBot, your SynData Arena assistant. "
            "Ask me about your uploaded dataset."
        )

    if any(word in question for word in ["shape", "size", "records", "rows"]):
        return (
            f"Your dataset contains {rows} records "
            f"and {cols} columns."
        )

    if "column" in question or "feature" in question:
        return (
            "Your dataset columns are:\n"
            + "\n".join(f"- {col}" for col in df.columns)
        )

    if "missing" in question or "null" in question:
        missing = df.isnull().sum()
        missing = missing[missing > 0]

        if missing.empty:
            return "There are no missing values in your dataset."

        return (
            "Missing values by column:\n"
            + "\n".join(
                f"- {col}: {count}"
                for col, count in missing.items()
            )
        )

    if "duplicate" in question:
        return (
            f"Your dataset contains "
            f"{df.duplicated().sum()} duplicate records."
        )

    if any(word in question for word in ["average", "mean"]):
        if numerical.empty:
            return "There are no numerical columns to calculate averages."

        return (
            "Column averages:\n"
            + "\n".join(
                f"- {col}: {df[col].mean():.2f}"
                for col in numerical.columns
                if df[col].notna().any()
            )
        )

    if "correlation" in question:
        if len(numerical.columns) < 2:
            return "At least two numerical columns are needed for correlation."

        return "Correlation matrix:\n" + numerical.corr().round(2).to_string()

    if "summary" in question or "analyse" in question or "analyze" in question:
        return generate_python_insights(df)

    if "categorical" in question:
        return (
            "Categorical columns:\n"
            + "\n".join(f"- {col}" for col in categorical.columns)
        )

    if "numerical" in question or "numeric" in question:
        return (
            "Numerical columns:\n"
            + "\n".join(f"- {col}" for col in numerical.columns)
        )

    return (
        "I can help you explore your dataset. Try asking:\n"
        "- What is the dataset size?\n"
        "- Show all columns.\n"
        "- Are there missing values?\n"
        "- Calculate column averages.\n"
        "- Show correlations.\n"
        "- Give me a dataset summary."
    )


# ============================================
# COMPATIBILITY FUNCTION
# Keeps old calls from breaking
# ============================================

def gemini_generate(prompt, image=None, retries=1):

    return (
        "Gemini has been replaced by the Python Intelligent "
        "Analysis Engine. Use generate_python_insights(df) "
        "for dataset insights."
    )


def test_gemini_connection():

    return (
        "Gemini is disabled. "
        "SynData Arena is using its Python Intelligent Analysis Engine."
    )


# ============================================
# INITIALIZATION STATUS
# ============================================

print("\n====================================")
print("       SYNDATA ARENA INITIALIZED")
print("====================================")
print("AI Engine:", AI_ENGINE)
print("Gemini Dependency: DISABLED")
print("Local Image AI: BLIP (Cell 3)")
print("Dataset Insights: Python")
print("Chatbot: Lightweight SynBot")
print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("====================================")
```

```python

# ============================================================
# SYNDATA ARENA - CELL 3
# CSV + LOCAL BLIP AI IMAGE PROCESSING
# TRANSFORMERS 5 COMPATIBLE
# NO GEMINI API REQUIRED
# ============================================================

import os
import re
import warnings

import numpy as np
import pandas as pd
import torch
from PIL import Image
from transformers import pipeline

warnings.filterwarnings("ignore")

# ============================================================
# SHARED DATASET VARIABLES
# ============================================================

uploaded_df = None
original_df = None
uploaded_image = None
uploaded_csv_name = None
uploaded_image_name = None
image_extraction_status = None

blip_captioner = None


# ============================================================
# LOAD BLIP IMAGE CAPTIONING MODEL
# ============================================================

def load_blip_models():

    global blip_captioner

    if blip_captioner is not None:
        return

    device = 0 if torch.cuda.is_available() else -1

    print("Loading BLIP image captioning model...")
    print("This may take a few minutes the first time.")

    blip_captioner = pipeline(
        "image-to-text",
        model="Salesforce/blip-image-captioning-base",
        device=device
    )

    print("BLIP image captioning model loaded!")


# ============================================================
# SHARED DATASET STORAGE
# ============================================================

def set_shared_dataset(new_df):

    global uploaded_df, original_df
    global df, dataset, data
    global original_data, uploaded_data

    uploaded_df = new_df.copy()
    original_df = new_df.copy()

    df = new_df.copy()
    dataset = new_df.copy()
    data = new_df.copy()
    original_data = new_df.copy()
    uploaded_data = new_df.copy()


def get_uploaded_dataset():

    if isinstance(uploaded_df, pd.DataFrame):
        if not uploaded_df.empty:
            return uploaded_df.copy()

    return None


def get_current_dataset():

    return get_uploaded_dataset()


# ============================================================
# CSV UPLOAD
# ============================================================

def upload_csv(file_path):

    global uploaded_image
    global uploaded_image_name
    global uploaded_csv_name
    global image_extraction_status

    if file_path is None:
        return "Please select a CSV file.", None

    try:

        if isinstance(file_path, str):
            path = file_path
        else:
            path = getattr(
                file_path,
                "name",
                getattr(file_path, "path", str(file_path))
            )

        if not path.lower().endswith(".csv"):
            return "Please upload a CSV file.", None

        try:
            new_df = pd.read_csv(path)
        except UnicodeDecodeError:
            new_df = pd.read_csv(path, encoding="latin-1")

        if new_df.empty:
            return "The uploaded CSV is empty.", None

        set_shared_dataset(new_df)

        uploaded_csv_name = os.path.basename(path)

        uploaded_image = None
        uploaded_image_name = None
        image_extraction_status = None

        status = (
            "CSV UPLOADED SUCCESSFULLY!\n\n"
            f"File: {uploaded_csv_name}\n"
            f"Records: {new_df.shape[0]}\n"
            f"Columns: {new_df.shape[1]}\n\n"
            "Dataset is ready for all SynData Arena modules."
        )

        return status, new_df.head(10)

    except Exception as e:

        return f"CSV upload error: {str(e)}", None


# ============================================================
# CONVERT INPUT TO PIL IMAGE
# ============================================================

def convert_to_pil(image):

    if isinstance(image, Image.Image):
        return image.copy()

    if isinstance(image, np.ndarray):
        return Image.fromarray(image.astype("uint8"))

    if isinstance(image, str):
        return Image.open(image)

    if hasattr(image, "name"):
        return Image.open(image.name)

    if hasattr(image, "path"):
        return Image.open(image.path)

    raise ValueError("Unsupported image format.")


# ============================================================
# IDENTIFY CATEGORY FROM IMAGE CAPTION
# ============================================================

def identify_category(text):

    name = str(text).lower()

    categories = {
        "Animal": [
            "dog", "cat", "bird", "horse", "cow",
            "lion", "tiger", "elephant", "fish",
            "deer", "monkey", "rabbit"
        ],
        "Nature": [
            "tree", "forest", "plant", "flower",
            "grass", "mountain", "lake", "river",
            "waterfall", "ocean", "sea", "sky",
            "cloud", "rock", "beach"
        ],
        "Vehicle": [
            "car", "bus", "truck", "motorcycle",
            "bicycle", "train", "airplane", "boat"
        ],
        "Building": [
            "building", "house", "bridge", "tower",
            "office", "school", "hotel"
        ],
        "Furniture": [
            "chair", "table", "sofa", "bed",
            "desk", "cabinet"
        ],
        "Food": [
            "food", "fruit", "apple", "banana",
            "bread", "pizza", "vegetable"
        ],
        "Person": [
            "person", "people", "man", "woman",
            "child", "boy", "girl"
        ]
    }

    for category, keywords in categories.items():

        if any(word in name for word in keywords):
            return category

    return "Other"


# ============================================================
# GET DOMINANT COLOUR DESCRIPTION
# ============================================================

def get_dominant_colors(image):

    small_image = image.resize((100, 100))

    pixels = np.asarray(
        small_image,
        dtype=np.float32
    )

    mean_rgb = pixels.mean(axis=(0, 1))

    r, g, b = mean_rgb

    if r > g and r > b:
        dominant = "Red"
    elif g > r and g > b:
        dominant = "Green"
    elif b > r and b > g:
        dominant = "Blue"
    else:
        dominant = "Mixed"

    return dominant, float(r), float(g), float(b)


# ============================================================
# IMAGE DATASET EXTRACTION
# ============================================================

def extract_image_dataset(image):

    image = convert_to_pil(image).convert("RGB")

    load_blip_models()

    print("Generating image caption...")

    caption_result = blip_captioner(image)

    if not caption_result:
        raise ValueError("BLIP could not describe the image.")

    caption = str(
        caption_result[0]["generated_text"]
    ).strip()

    if not caption:
        raise ValueError("Empty image caption received.")

    print("Image caption:", caption)

    width, height = image.size

    dominant, red, green, blue = get_dominant_colors(image)

    brightness = float(
        np.asarray(image.resize((100, 100)), dtype=np.float32).mean()
    )

    category = identify_category(caption)

    # Extract a short subject description from the caption
    subject = re.sub(
        r"^(a|an|the)\s+",
        "",
        caption,
        flags=re.IGNORECASE
    ).strip()

    # Create structured dataset
    extracted_df = pd.DataFrame([{

        "image_name": uploaded_image_name or "Uploaded_Image",

        "main_subject": subject,

        "object_category": category,

        "image_caption": caption,

        "image_width": width,

        "image_height": height,

        "image_area": width * height,

        "average_brightness": round(brightness, 2),

        "average_red": round(red, 2),

        "average_green": round(green, 2),

        "average_blue": round(blue, 2),

        "dominant_color": dominant,

        "source": "BLIP Local Vision"

    }])

    description = (
        f"Image caption: {caption}\n"
        f"Main subject: {subject}\n"
        f"Category: {category}\n"
        f"Dominant colour: {dominant}\n"
        f"Image dimensions: {width} x {height}"
    )

    return extracted_df, description, "General Image / Photograph"


# ============================================================
# IMAGE UPLOAD
# ============================================================

def upload_image(image):

    global uploaded_image
    global uploaded_image_name
    global uploaded_csv_name
    global image_extraction_status

    if image is None:
        return "Please select an image.", None

    try:

        uploaded_image = convert_to_pil(image).convert("RGB")

        original_format = uploaded_image.format or "Unknown"

        uploaded_image_name = "Uploaded_Image"
        uploaded_csv_name = "Image_Extracted_Dataset"

        print("========================================")
        print("STARTING BLIP IMAGE PROCESSING")
        print("========================================")

        extracted_df, description, data_type = (
            extract_image_dataset(uploaded_image)
        )

        if extracted_df is None or extracted_df.empty:
            raise ValueError("No meaningful data was extracted.")

        set_shared_dataset(extracted_df)

        image_extraction_status = {

            "description": description,
            "data_type": data_type,
            "rows": len(extracted_df),
            "columns": len(extracted_df.columns),
            "processing": "BLIP Local AI"

        }

        status = (
            "IMAGE ANALYZED SUCCESSFULLY!\n\n"
            f"Image dimensions: {uploaded_image.size}\n"
            f"Image format: {original_format}\n"
            f"Image type: {data_type}\n\n"
            "LOCAL AI IMAGE UNDERSTANDING\n"
            f"Dataset rows: {len(extracted_df)}\n"
            f"Dataset columns: {len(extracted_df.columns)}\n\n"
            f"Image analysis:\n{description}\n\n"
            "The extracted dataset is stored for the "
            "SynData Arena modules."
        )

        print("Extracted dataset shape:", extracted_df.shape)
        print("Extracted columns:", list(extracted_df.columns))

        return status, extracted_df.head(10)

    except Exception as e:

        print("IMAGE PROCESSING ERROR:", str(e))

        image_extraction_status = {
            "error": str(e)
        }

        return (
            "IMAGE PROCESSING ERROR\n\n"
            f"{str(e)}",
            None
        )


# ============================================================
# READY
# ============================================================

print("================================================")
print("       SYNDATA ARENA BLIP IMAGE PROCESSING")
print("================================================")
print("CSV Upload: READY")
print("BLIP Image Captioning: READY")
print("Image Feature Extraction: READY")
print("Gemini Image API: NOT REQUIRED")
print("Shared Dataset Storage: READY")
print("================================================")
```

```python

# ============================================================
# SYNDATA ARENA - CELL 4
# ADVANCED DATASET ANALYSIS ENGINE
# PYTHON-POWERED AI INSIGHTS
# NO GEMINI DEPENDENCY
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import io
import json
import traceback

# ============================================================
# DATASET ANALYSIS FUNCTION
# ============================================================

def analyze_dataset():

    global uploaded_df

    if uploaded_df is None or uploaded_df.empty:
        return (
            "Please upload a CSV dataset first.",
            None, None, None, None, None, None, None
        )

    try:

        df = uploaded_df.copy()

        # ----------------------------------------
        # 1. BASIC DATASET INFORMATION
        # ----------------------------------------

        rows, columns = df.shape

        overview = pd.DataFrame({
            "Metric": [
                "Total Rows",
                "Total Columns",
                "Total Numerical Columns",
                "Total Categorical Columns",
                "Total Missing Values",
                "Total Duplicate Rows",
                "Memory Usage (KB)"
            ],
            "Value": [
                rows,
                columns,
                len(df.select_dtypes(include=np.number).columns),
                len(df.select_dtypes(
                    include=["object", "category", "bool"]
                ).columns),
                int(df.isnull().sum().sum()),
                int(df.duplicated().sum()),
                round(df.memory_usage(deep=True).sum() / 1024, 2)
            ]
        })

        # ----------------------------------------
        # 2. COLUMN INFORMATION
        # ----------------------------------------

        column_info = pd.DataFrame({
            "Column Name": df.columns.astype(str),
            "Data Type": df.dtypes.astype(str).values,
            "Non-Null Count": df.notnull().sum().values,
            "Missing Count": df.isnull().sum().values,
            "Missing Percentage": (
                df.isnull().mean().values * 100
            ).round(2)
        })

        # ----------------------------------------
        # 3. NUMERICAL STATISTICS
        # ----------------------------------------

        numerical = df.select_dtypes(include=np.number)

        if not numerical.empty:

            numerical_stats = numerical.describe().T.round(3)
            numerical_stats = numerical_stats.reset_index()

            numerical_stats.rename(
                columns={"index": "Column"},
                inplace=True
            )

        else:

            numerical_stats = pd.DataFrame({
                "Information": ["No numerical columns found."]
            })

        # ----------------------------------------
        # 4. CATEGORICAL ANALYSIS
        # ----------------------------------------

        categorical = df.select_dtypes(
            include=["object", "category", "bool"]
        )

        categorical_results = []

        for col in categorical.columns:

            unique_count = df[col].nunique(dropna=True)

            top_value = df[col].mode(dropna=True)

            most_common = (
                str(top_value.iloc[0])
                if not top_value.empty else "No data"
            )

            categorical_results.append({
                "Column": str(col),
                "Unique Values": int(unique_count),
                "Most Frequent Value": most_common
            })

        categorical_stats = pd.DataFrame(categorical_results)

        if categorical_stats.empty:

            categorical_stats = pd.DataFrame({
                "Information": ["No categorical columns found."]
            })

        # ----------------------------------------
        # 5. MISSING VALUE ANALYSIS
        # ----------------------------------------

        missing = df.isnull().sum()

        missing_stats = pd.DataFrame({
            "Column": missing.index.astype(str),
            "Missing Values": missing.values,
            "Missing Percentage": (
                missing.values / max(rows, 1) * 100
            ).round(2)
        })

        missing_stats = missing_stats[
            missing_stats["Missing Values"] > 0
        ]

        # ----------------------------------------
        # 6. DUPLICATE ANALYSIS
        # ----------------------------------------

        duplicate_count = int(df.duplicated().sum())

        # ----------------------------------------
        # 7. CORRELATION ANALYSIS
        # ----------------------------------------

        if numerical.shape[1] >= 2:

            correlation = numerical.corr().round(3)

            fig, ax = plt.subplots(figsize=(9, 6))

            sns.heatmap(
                correlation,
                annot=True,
                cmap="coolwarm",
                fmt=".2f",
                ax=ax
            )

            ax.set_title("Dataset Correlation Heatmap")

            plt.tight_layout()

            correlation_plot = fig

        else:

            correlation_plot = None

        # ----------------------------------------
        # 8. AUTOMATIC NUMERICAL DISTRIBUTIONS
        # ----------------------------------------

        if not numerical.empty:

            plot_columns = list(numerical.columns[:6])

            fig2, axes = plt.subplots(
                len(plot_columns),
                1,
                figsize=(9, 4 * len(plot_columns))
            )

            if len(plot_columns) == 1:
                axes = [axes]

            for ax, col in zip(axes, plot_columns):

                sns.histplot(
                    df[col].dropna(),
                    kde=True,
                    ax=ax
                )

                ax.set_title(f"Distribution of {col}")

            plt.tight_layout()

            distribution_plot = fig2

        else:

            distribution_plot = None

        # ----------------------------------------
        # 9. AUTOMATIC PYTHON AI DATASET INSIGHTS
        # ----------------------------------------

        insight_text = generate_python_insights(df)

        # ----------------------------------------
        # 10. FINAL REPORT
        # ----------------------------------------

        report = f"""
SYNDATA ARENA - DATASET ANALYSIS REPORT

DATASET OVERVIEW
----------------
Total Rows: {rows}
Total Columns: {columns}
Numerical Columns: {len(numerical.columns)}
Categorical Columns: {len(categorical.columns)}
Missing Values: {int(df.isnull().sum().sum())}
Duplicate Rows: {duplicate_count}

AUTOMATIC PYTHON AI INSIGHTS
----------------------------
{insight_text}
"""

        # ----------------------------------------
        # 11. RETURN RESULTS
        # ----------------------------------------

        return (
            report,
            overview,
            column_info,
            numerical_stats,
            categorical_stats,
            missing_stats,
            correlation_plot,
            distribution_plot
        )

    except Exception as e:

        print("Dataset analysis error:", str(e))
        traceback.print_exc()

        return (
            f"Analysis failed: {str(e)}",
            None, None, None, None, None, None, None
        )


# ============================================================
# INITIALIZATION
# ============================================================

print("SynData Arena Dataset Analysis Engine created!")
print("AI Insights: Python Intelligent Analysis Engine")
print("Gemini Dependency: DISABLED")
print("CSV Analysis: READY")
print("Correlation Analysis: READY")
print("Numerical Distribution Analysis: READY")
```

```python

# ============================================================
# SYNDATA ARENA - CELL 6
# SYNBOT AI - INTELLIGENT DATA ANALYST
# PYTHON + PANDAS | NO GEMINI DEPENDENCY
# ============================================================

import pandas as pd
import numpy as np
import json
import re
import warnings

warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# 1. CREATE DATASET PROFILE
# ------------------------------------------------------------

def synbot_dataset_profile(df):

    numeric = df.select_dtypes(include=np.number)

    categorical = df.select_dtypes(
        include=["object", "category", "bool"]
    )

    profile = {
        "rows": int(len(df)),
        "columns": int(len(df.columns)),
        "column_names": [str(c) for c in df.columns],
        "data_types": {
            str(c): str(t) for c, t in df.dtypes.items()
        },
        "missing_values": {
            str(c): int(v)
            for c, v in df.isnull().sum().items()
        },
        "duplicate_rows": int(df.duplicated().sum()),
        "numeric_summary": {},
        "categorical_summary": {},
        "sample_records": df.head(5).fillna("").astype(str).to_dict(
            orient="records"
        )
    }

    if not numeric.empty:
        profile["numeric_summary"] = (
            numeric.describe()
            .replace([np.inf, -np.inf], np.nan)
            .round(3)
            .fillna("")
            .to_dict()
        )

    for col in categorical.columns[:25]:
        profile["categorical_summary"][str(col)] = (
            categorical[col]
            .dropna()
            .astype(str)
            .value_counts()
            .head(5)
            .to_dict()
        )

    return profile


# ------------------------------------------------------------
# 2. IDENTIFY COLUMNS FROM USER QUESTION
# ------------------------------------------------------------

def synbot_find_columns(question, df):

    question_lower = question.lower()

    matched = []

    # Match longer column names first
    sorted_columns = sorted(
        df.columns,
        key=lambda x: len(str(x)),
        reverse=True
    )

    for col in sorted_columns:

        col_text = str(col).lower()

        if col_text in question_lower:
            matched.append(col)

    return matched


# ------------------------------------------------------------
# 3. UNDERSTAND QUESTION USING PYTHON RULES
# ------------------------------------------------------------

def synbot_create_analysis_plan(question, profile):

    question = str(question).strip()
    q = question.lower()

    columns = profile["column_names"]

    numeric_cols = [
        col for col, dtype in profile["data_types"].items()
        if any(x in dtype.lower() for x in ["int", "float"])
    ]

    categorical_cols = [
        col for col, dtype in profile["data_types"].items()
        if any(x in dtype.lower() for x in ["object", "category", "bool"])
    ]

    matched = []

    for col in sorted(columns, key=len, reverse=True):

        if str(col).lower() in q:
            matched.append(col)

    plan = {
        "operation": "general",
        "columns": matched,
        "statistic": "mean",
        "group_by": [],
        "filters": [],
        "top_n": 10,
        "explanation": "General dataset analysis",
        "general_answer": ""
    }

    # Detect requested number of results
    number_match = re.search(
        r"(?:top|first|show|display)\s+(\d+)",
        q
    )

    if number_match:
        plan["top_n"] = min(int(number_match.group(1)), 100)

    # Detect requested statistic
    if any(x in q for x in ["median", "middle value"]):
        plan["statistic"] = "median"

    elif any(x in q for x in ["maximum", "highest", "largest", "max"]):
        plan["statistic"] = "max"

    elif any(x in q for x in ["minimum", "lowest", "smallest", "min"]):
        plan["statistic"] = "min"

    elif any(x in q for x in ["total", "sum", "overall amount"]):
        plan["statistic"] = "sum"

    elif any(x in q for x in ["count", "number of"]):
        plan["statistic"] = "count"

    elif any(x in q for x in ["standard deviation", "std"]):
        plan["statistic"] = "std"

    else:
        plan["statistic"] = "mean"

    # Detect analysis operation
    if any(x in q for x in ["missing", "null", "empty values"]):

        plan["operation"] = "missing"
        plan["explanation"] = "Checking missing values."

    elif "duplicate" in q:

        plan["operation"] = "duplicates"
        plan["explanation"] = "Checking duplicate records."

    elif "correlation" in q or "relationship between" in q:

        plan["operation"] = "correlation"
        plan["explanation"] = "Calculating numerical correlations."

    elif "outlier" in q or "anomal" in q:

        plan["operation"] = "outliers"
        plan["explanation"] = "Detecting potential outliers."

    elif "unique" in q or "distinct" in q:

        plan["operation"] = "unique"
        plan["explanation"] = "Finding unique values."

    elif any(x in q for x in ["distribution", "frequency distribution"]):

        plan["operation"] = "distribution"
        plan["explanation"] = "Examining value distribution."

    elif any(x in q for x in ["value counts", "most common", "most frequent"]):

        plan["operation"] = "value_counts"
        plan["explanation"] = "Finding the most frequent values."

    elif "group" in q or "grouped" in q or "by category" in q:

        plan["operation"] = "groupby"

        if matched:
            plan["group_by"] = matched[:1]

        elif categorical_cols:
            plan["group_by"] = categorical_cols[:1]

        plan["explanation"] = "Comparing dataset groups."

    elif any(x in q for x in ["filter", "greater than", "less than", "above", "below"]):

        plan["operation"] = "filter"
        plan["explanation"] = "Filtering dataset records."

    elif any(x in q for x in ["average", "mean", "median", "maximum", "minimum", "highest", "lowest", "sum", "total"]):

        if matched:
            plan["operation"] = "column_stats"
            plan["explanation"] = (
                f"Calculating {plan['statistic']} for {matched[0]}."
            )
        else:
            plan["operation"] = "column_stats"
            plan["columns"] = numeric_cols[:1]
            plan["explanation"] = "Calculating numerical statistics."

    elif any(x in q for x in ["describe", "statistics", "summary", "analyse", "analyze", "overview"]):

        plan["operation"] = "describe"
        plan["explanation"] = "Generating dataset summary."

    elif "column" in q or "feature" in q:

        plan["operation"] = "columns"
        plan["explanation"] = "Listing dataset columns."

    elif "row" in q or "record" in q or "dataset size" in q:

        plan["operation"] = "overview"
        plan["explanation"] = "Showing dataset size."

    return plan


# ------------------------------------------------------------
# 4. VALIDATE COLUMN NAMES
# ------------------------------------------------------------

def synbot_validate_columns(columns, df):

    if not isinstance(columns, list):
        columns = []

    return [
        col for col in columns
        if col in df.columns
    ]


# ------------------------------------------------------------
# 5. EXECUTE ANALYSIS USING PYTHON
# ------------------------------------------------------------

def synbot_execute_plan(plan, df):

    operation = str(
        plan.get("operation", "general")
    ).lower()

    columns = synbot_validate_columns(
        plan.get("columns", []), df
    )

    numeric_cols = df.select_dtypes(
        include=np.number
    ).columns.tolist()

    # Dataset overview
    if operation == "overview":

        return (
            f"Total records: {len(df)}\n"
            f"Total columns: {len(df.columns)}\n"
            f"Numerical columns: {len(numeric_cols)}\n"
            f"Missing values: {df.isnull().sum().sum()}\n"
            f"Duplicate records: {df.duplicated().sum()}"
        )

    # Column names
    if operation == "columns":

        return "Dataset columns:\n" + "\n".join(
            f"- {col}" for col in df.columns
        )

    # Dataset description
    if operation == "describe":

        return df.describe(include="all").to_string()

    # Numerical statistics
    if operation == "column_stats":

        if not columns:
            return "Please mention a valid numerical column name."

        col = columns[0]

        values = pd.to_numeric(
            df[col], errors="coerce"
        ).replace([np.inf, -np.inf], np.nan).dropna()

        if values.empty:
            return f"No numerical values found in {col}."

        stat = str(
            plan.get("statistic", "mean")
        ).lower()

        functions = {
            "mean": values.mean,
            "median": values.median,
            "sum": values.sum,
            "min": values.min,
            "max": values.max,
            "std": values.std,
            "count": values.count
        }

        if stat not in functions:
            stat = "mean"

        return (
            f"Column: {col}\n"
            f"Statistic: {stat}\n"
            f"Result: {functions[stat]():.3f}\n"
            f"Valid records: {len(values)}"
        )

    # Value counts
    if operation == "value_counts":

        if not columns:
            columns = df.select_dtypes(
                include=["object", "category", "bool"]
            ).columns.tolist()[:1]

        if not columns:
            return "No suitable column was found."

        return (
            df[columns[0]]
            .value_counts(dropna=False)
            .head(plan.get("top_n", 10))
            .to_string()
        )

    # Group comparison
    if operation == "groupby":

        group_cols = synbot_validate_columns(
            plan.get("group_by", []), df
        )

        if not group_cols:
            group_cols = columns[:1]

        if not group_cols:
            group_cols = df.select_dtypes(
                include=["object", "category", "bool"]
            ).columns.tolist()[:1]

        if not group_cols:
            return "No suitable grouping column was found."

        metric = None

        for col in columns:
            if col not in group_cols:
                metric = col
                break

        if metric and metric in numeric_cols:

            stat = plan.get("statistic", "mean")

            if stat not in ["mean", "sum", "min", "max", "count", "median"]:
                stat = "mean"

            result = df.groupby(
                group_cols, dropna=False
            )[metric].agg(stat)

        else:

            result = df.groupby(
                group_cols, dropna=False
            ).size()

        return result.sort_values(
            ascending=False
        ).head(plan.get("top_n", 20)).to_string()

    # Correlation
    if operation == "correlation":

        selected = [
            c for c in columns if c in numeric_cols
        ]

        if len(selected) < 2:
            selected = numeric_cols

        if len(selected) < 2:
            return "At least two numerical columns are required."

        return df[selected].corr().round(3).to_string()

    # Missing values
    if operation == "missing":

        result = df.isnull().sum()
        result = result[result > 0].sort_values(
            ascending=False
        )

        if result.empty:
            return "No missing values were found."

        return result.to_string()

    # Duplicate records
    if operation == "duplicates":

        return (
            f"Total records: {len(df)}\n"
            f"Duplicate records: {df.duplicated().sum()}\n"
            f"Unique records: {len(df.drop_duplicates())}"
        )

    # Outlier detection
    if operation == "outliers":

        selected = [
            c for c in columns if c in numeric_cols
        ]

        if not selected:
            selected = numeric_cols

        if not selected:
            return "No numerical columns are available."

        output = {}

        for col in selected:

            values = pd.to_numeric(
                df[col], errors="coerce"
            ).dropna()

            if len(values) < 4:
                continue

            q1 = values.quantile(0.25)
            q3 = values.quantile(0.75)
            iqr = q3 - q1

            lower = q1 - 1.5 * iqr
            upper = q3 + 1.5 * iqr

            count = int(
                ((values < lower) | (values > upper)).sum()
            )

            output[col] = {
                "outliers": count,
                "lower_bound": float(lower),
                "upper_bound": float(upper)
            }

        return json.dumps(output, indent=2)

    # Filter records
    if operation == "filter":

        if not columns:
            return (
                "Please mention a valid column name "
                "when asking to filter records."
            )

        col = columns[0]

        number_match = re.search(
            r"(?:greater than|above|over|more than|less than|below|under)\s+(-?\d+(?:\.\d+)?)",
            str(plan.get("question", "")).lower()
        )

        if not number_match:
            return (
                "Please specify a filter condition, "
                "for example: show records where Age is greater than 25."
            )

        value = float(number_match.group(1))
        q = plan.get("question", "").lower()

        series = pd.to_numeric(df[col], errors="coerce")

        if any(x in q for x in ["less than", "below", "under"]):
            result = df[series < value]
        else:
            result = df[series > value]

        return (
            f"Matching records: {len(result)}\n\n"
            + result.head(30).to_string(index=False)
        )

    # Unique values
    if operation == "unique":

        if not columns:
            return "Please mention a valid column name."

        col = columns[0]

        return (
            f"Column: {col}\n"
            f"Unique count: {df[col].nunique(dropna=True)}\n"
            f"Values:\n{df[col].dropna().unique()[:50].tolist()}"
        )

    # Distribution
    if operation == "distribution":

        if not columns:
            return "Please mention a valid column name."

        col = columns[0]

        if col in numeric_cols:
            return df[col].describe().to_string()

        return df[col].value_counts().head(20).to_string()

    # General analysis
    if operation == "general":

        return generate_python_insights(df)

    return "I could not identify the requested analysis."


# ------------------------------------------------------------
# 6. MAIN SYNBOT FUNCTION
# ------------------------------------------------------------

def synbot_ai(user_question):

    global uploaded_df

    if uploaded_df is None or uploaded_df.empty:
        return (
            "Please upload a CSV dataset or image before "
            "asking questions."
        )

    if not isinstance(user_question, str) or not user_question.strip():
        return "Please enter a question."

    try:

        df = uploaded_df.copy()

        profile = synbot_dataset_profile(df)

        plan = synbot_create_analysis_plan(
            user_question,
            profile
        )

        # Pass original question to filtering logic
        plan["question"] = user_question

        result = synbot_execute_plan(plan, df)

        return f"""
## SynBot AI - Intelligent Data Analyst

**Your question:** {user_question}

**Analysis:** {plan.get("explanation", "Dataset analysis")}

### Results

{result}

### Dataset Context

- Total records: {len(df):,}
- Total columns: {len(df.columns)}
- Numerical columns: {len(df.select_dtypes(include=np.number).columns)}
- Analysis engine: Python + Pandas
- Gemini dependency: Disabled

*Results are calculated directly from your uploaded dataset.*
"""

    except Exception as e:

        print("SynBot error:", str(e))

        return (
            "SynBot encountered an error while analysing "
            "your dataset: " + str(e)
        )


# ------------------------------------------------------------
# 7. INITIALIZATION
# ------------------------------------------------------------

print("================================================")
print("       SYNBOT AI - DATA ANALYST READY")
print("================================================")
print("Dataset Profiling: READY")
print("Natural Language Question Matching: READY")
print("Python and Pandas Analysis: READY")
print("Statistical Analysis: READY")
print("Group Comparison: READY")
print("Correlation Analysis: READY")
print("Outlier Detection: READY")
print("Dataset Filtering: READY")
print("Gemini Dependency: DISABLED")
print("Local SynBot: READY")
print("================================================")
```

```python

# ============================================================
# SYNDATA ARENA - CELL 8
# ADVANCED SYNTHETIC DATA GENERATION ENGINE
# Minimum: 1 | Maximum: 10,000
# ============================================================

import pandas as pd
import numpy as np
import os
import tempfile
import traceback

# Store latest generated dataset
synthetic_df = None
synthetic_file_path = None
synthetic_generation_summary = None


# ============================================================
# SYNTHETIC DATA GENERATION FUNCTION
# ============================================================

def generate_synthetic_data(number_of_records):

    global uploaded_df
    global synthetic_df
    global synthetic_file_path
    global synthetic_generation_summary

    try:

        # ----------------------------------------------------
        # 1. VALIDATE ORIGINAL DATASET
        # ----------------------------------------------------

        if uploaded_df is None:
            raise ValueError(
                "Please upload a CSV dataset first."
            )

        if uploaded_df.empty:
            raise ValueError(
                "The uploaded dataset is empty."
            )

        # Prevent image metadata from entering CSV generation
        image_status = globals().get(
            "image_extraction_status", {}
        )

        if isinstance(image_status, dict):

            processing_type = str(
                image_status.get("processing", "")
            ).upper()

            if "BLIP" in processing_type:
                raise ValueError(
                    "Synthetic data generation is available "
                    "for CSV datasets only. Please upload a "
                    "CSV dataset to use this feature."
                )

        original_df = uploaded_df.copy()

        # ----------------------------------------------------
        # 2. VALIDATE RECORD COUNT
        # ----------------------------------------------------

        number_of_records = int(number_of_records)

        if not 1 <= number_of_records <= 10000:
            raise ValueError(
                "Record count must be between 1 and 10,000."
            )

        rng = np.random.default_rng()

        original_rows = len(original_df)
        original_columns = len(original_df.columns)

        # ----------------------------------------------------
        # 3. BOOTSTRAP ROW SAMPLING
        # ----------------------------------------------------

        sampled_indices = rng.integers(
            low=0,
            high=original_rows,
            size=number_of_records
        )

        result_df = original_df.iloc[
            sampled_indices
        ].reset_index(drop=True).copy()

        # ----------------------------------------------------
        # 4. IDENTIFY COLUMN TYPES
        # ----------------------------------------------------

        numerical_columns = original_df.select_dtypes(
            include=np.number
        ).columns.tolist()

        categorical_columns = original_df.select_dtypes(
            include=["object", "category", "bool"]
        ).columns.tolist()

        datetime_columns = original_df.select_dtypes(
            include=["datetime", "datetimetz"]
        ).columns.tolist()

        modified_columns = []

        # ----------------------------------------------------
        # 5. GENERATE NUMERICAL VARIATIONS
        # ----------------------------------------------------

        for col in numerical_columns:

            source = pd.to_numeric(
                original_df[col],
                errors="coerce"
            )

            valid_values = source.dropna()

            # Skip constant or binary columns
            if len(valid_values) < 2:
                continue

            if valid_values.nunique() <= 2:
                continue

            std_value = valid_values.std()

            if pd.isna(std_value) or std_value == 0:
                continue

            min_value = valid_values.min()
            max_value = valid_values.max()

            generated_values = pd.to_numeric(
                result_df[col],
                errors="coerce"
            ).to_numpy(dtype=float)

            valid_mask = ~np.isnan(generated_values)

            noise = rng.normal(
                loc=0,
                scale=std_value * 0.03,
                size=number_of_records
            )

            generated_values[valid_mask] += noise[valid_mask]

            generated_values[valid_mask] = np.clip(
                generated_values[valid_mask],
                min_value,
                max_value
            )

            # Preserve integer data
            if pd.api.types.is_integer_dtype(original_df[col]):
                generated_values[valid_mask] = np.rint(
                    generated_values[valid_mask]
                )

            result_df[col] = generated_values

            modified_columns.append(str(col))

        # ----------------------------------------------------
        # 6. GENERATE NEW IDENTIFIERS
        # ----------------------------------------------------

        for col in original_df.columns:

            column_name = str(col).lower()

            is_id_column = (
                column_name == "id"
                or column_name.endswith("_id")
                or column_name.endswith("id")
            )

            if not is_id_column:
                continue

            if pd.api.types.is_integer_dtype(original_df[col]):

                valid_ids = pd.to_numeric(
                    original_df[col],
                    errors="coerce"
                ).dropna()

                if (
                    len(valid_ids) > 0
                    and valid_ids.nunique() == len(valid_ids)
                ):

                    start_id = int(valid_ids.max()) + 1

                    result_df[col] = np.arange(
                        start_id,
                        start_id + number_of_records
                    )

                    modified_columns.append(str(col))

            elif (
                pd.api.types.is_object_dtype(original_df[col])
                or pd.api.types.is_string_dtype(original_df[col])
            ):

                result_df[col] = [
                    f"SYN_{i+1:06d}"
                    for i in range(number_of_records)
                ]

                modified_columns.append(str(col))

        # ----------------------------------------------------
        # 7. PRESERVE ORIGINAL COLUMN ORDER
        # ----------------------------------------------------

        result_df = result_df[original_df.columns]

        # ----------------------------------------------------
        # 8. GENERATE SUMMARY STATISTICS
        # ----------------------------------------------------

        missing_values = int(
            result_df.isnull().sum().sum()
        )

        duplicate_records = int(
            result_df.duplicated().sum()
        )

        unique_generated_records = int(
            result_df.drop_duplicates().shape[0]
        )

        numerical_count = len(numerical_columns)
        categorical_count = len(categorical_columns)
        datetime_count = len(datetime_columns)

        original_size = round(
            original_df.memory_usage(deep=True).sum() / 1024,
            2
        )

        generated_size = round(
            result_df.memory_usage(deep=True).sum() / 1024,
            2
        )

        # ----------------------------------------------------
        # 9. CREATE DOWNLOADABLE CSV
        # ----------------------------------------------------

        output_directory = tempfile.gettempdir()

        output_path = os.path.join(
            output_directory,
            "SynData_Arena_Synthetic_Data.csv"
        )

        result_df.to_csv(
            output_path,
            index=False
        )

        # ----------------------------------------------------
        # 10. CREATE BRIEF GENERATION SUMMARY
        # ----------------------------------------------------

        summary = f"""
================================================
           SYNDATA ARENA
       SYNTHETIC DATA GENERATION REPORT
================================================

GENERATION STATUS
-----------------
Successfully Generated

ORIGINAL DATASET
----------------
Original Records             : {original_rows:,}
Original Columns             : {original_columns}

SYNTHETIC DATASET
-----------------
Generated Records            : {number_of_records:,}
Generated Columns            : {original_columns}
Numerical Columns             : {numerical_count}
Categorical Columns           : {categorical_count}
Date/Time Columns             : {datetime_count}
Modified Columns              : {len(set(modified_columns))}
Unique Generated Records      : {unique_generated_records:,}
Duplicate Records             : {duplicate_records:,}
Missing Values                : {missing_values:,}

GENERATION METHOD
-----------------
Bootstrap Sampling
Controlled Numerical Variation
Fresh Identifier Generation

DATASET SIZE
------------
Original Dataset              : {original_size} KB
Synthetic Dataset              : {generated_size} KB

OUTPUT FILE
-----------
SynData_Arena_Synthetic_Data.csv

================================================
Synthetic data generation completed successfully!
================================================
"""

        # ----------------------------------------------------
        # 11. SAVE GENERATED DATASET
        # ----------------------------------------------------

        synthetic_df = result_df.copy()
        synthetic_file_path = output_path
        synthetic_generation_summary = summary

        return (
            result_df,
            summary,
            output_path
        )

    except Exception as e:

        traceback.print_exc()

        synthetic_df = None
        synthetic_file_path = None
        synthetic_generation_summary = None

        return (
            None,
            f"ERROR DURING SYNTHETIC DATA GENERATION:\n{str(e)}",
            None
        )


# ============================================================
# CELL 8 INITIALIZATION
# ============================================================

print("================================================")
print("       SYNDATA ARENA INITIALIZED")
print("================================================")
print("Synthetic Data Generation Engine: READY")
print("Minimum Records: 1")
print("Maximum Records: 10,000")
print("Numerical Data Generation: READY")
print("Categorical Data Handling: READY")
print("Identifier Generation: READY")
print("Missing Value Handling: READY")
print("CSV Download: READY")
print("Generation Summary: READY")
print("================================================")
```

```python

# ============================================================
# CELL 10: SYNTHETIC DATA QUALITY EVALUATION
# ============================================================

import pandas as pd
import numpy as np
import gradio as gr
from scipy.stats import ks_2samp


# ============================================================
# SYNTHETIC DATA QUALITY EVALUATION FUNCTION
# ============================================================

def evaluate_synthetic_quality():

    global uploaded_df, synthetic_df

    # --------------------------------------------------------
    # 1. VALIDATE DATASETS
    # --------------------------------------------------------

    if uploaded_df is None or uploaded_df.empty:
        return (
            "Please upload an original CSV dataset first.",
            pd.DataFrame()
        )

    image_status = globals().get(
        "image_extraction_status", {}
    )

    if isinstance(image_status, dict):

        processing_type = str(
            image_status.get("processing", "")
        ).upper()

        if "BLIP" in processing_type:
            return (
                "Quality evaluation is available for CSV datasets only. "
                "Please upload a CSV dataset.",
                pd.DataFrame()
            )

    if synthetic_df is None:
        return (
            "Please generate synthetic data using Cell 8 first.",
            pd.DataFrame()
        )

    original = uploaded_df.copy()
    synthetic = synthetic_df.copy()

    if original.empty or synthetic.empty:
        return (
            "Original or synthetic dataset is empty.",
            pd.DataFrame()
        )

    report = []
    results = []

    # --------------------------------------------------------
    # 2. BASIC DATASET INFORMATION
    # --------------------------------------------------------

    original_columns = list(original.columns)
    synthetic_columns = list(synthetic.columns)

    common_columns = [
        col for col in original_columns
        if col in synthetic_columns
    ]

    missing_columns = [
        col for col in original_columns
        if col not in synthetic_columns
    ]

    extra_columns = [
        col for col in synthetic_columns
        if col not in original_columns
    ]

    column_match = (
        len(common_columns) / len(original_columns)
        if original_columns else 0
    )

    report.append("# SYNTHETIC DATA QUALITY EVALUATION")
    report.append("## Dataset Overview")
    report.append(f"- Original records: **{len(original):,}**")
    report.append(f"- Synthetic records: **{len(synthetic):,}**")
    report.append(f"- Original columns: **{len(original_columns)}**")
    report.append(f"- Synthetic columns: **{len(synthetic_columns)}**")
    report.append(f"- Matching columns: **{len(common_columns)}**")

    if missing_columns:
        report.append(
            f"- Columns missing in synthetic data: **{', '.join(map(str, missing_columns))}**"
        )

    if extra_columns:
        report.append(
            f"- Additional synthetic columns: **{', '.join(map(str, extra_columns))}**"
        )

    if not common_columns:
        return (
            "No matching columns were found between the original "
            "and synthetic datasets.",
            pd.DataFrame()
        )

    # --------------------------------------------------------
    # 3. COLUMN DATA TYPE CONSISTENCY
    # --------------------------------------------------------

    type_matches = 0

    for col in common_columns:

        original_numeric = pd.api.types.is_numeric_dtype(
            original[col]
        )

        synthetic_numeric = pd.api.types.is_numeric_dtype(
            synthetic[col]
        )

        match = original_numeric == synthetic_numeric

        if match:
            type_matches += 1

        results.append({
            "Evaluation": f"Data type - {col}",
            "Original Value": str(original[col].dtype),
            "Synthetic Value": str(synthetic[col].dtype),
            "Result": "MATCH" if match else "MISMATCH"
        })

    type_score = (
        type_matches / len(common_columns)
        if common_columns else 0
    )

    # --------------------------------------------------------
    # 4. MISSING VALUE COMPARISON
    # --------------------------------------------------------

    missing_differences = []

    for col in common_columns:

        original_missing = original[col].isna().mean()
        synthetic_missing = synthetic[col].isna().mean()

        difference = abs(
            original_missing - synthetic_missing
        )

        missing_differences.append(difference)

        results.append({
            "Evaluation": f"Missing values - {col}",
            "Original Value": f"{original_missing * 100:.2f}%",
            "Synthetic Value": f"{synthetic_missing * 100:.2f}%",
            "Result": (
                "SIMILAR" if difference <= 0.05
                else "DIFFERENT"
            )
        })

    missing_score = (
        1 - np.mean(missing_differences)
        if missing_differences else 1
    )

    # --------------------------------------------------------
    # 5. DUPLICATE RECORD CHECK
    # --------------------------------------------------------

    original_duplicate_rate = original.duplicated().mean()
    synthetic_duplicate_rate = synthetic.duplicated().mean()

    duplicate_difference = abs(
        original_duplicate_rate - synthetic_duplicate_rate
    )

    duplicate_score = max(0, 1 - duplicate_difference)

    results.append({
        "Evaluation": "Duplicate record rate",
        "Original Value": f"{original_duplicate_rate * 100:.2f}%",
        "Synthetic Value": f"{synthetic_duplicate_rate * 100:.2f}%",
        "Result": (
            "SIMILAR" if duplicate_difference <= 0.05
            else "DIFFERENT"
        )
    })

    # --------------------------------------------------------
    # 6. NUMERICAL DISTRIBUTION COMPARISON
    # --------------------------------------------------------

    numerical_scores = []

    for col in common_columns:

        if (
            pd.api.types.is_numeric_dtype(original[col])
            and pd.api.types.is_numeric_dtype(synthetic[col])
        ):

            original_values = pd.to_numeric(
                original[col], errors="coerce"
            ).dropna()

            synthetic_values = pd.to_numeric(
                synthetic[col], errors="coerce"
            ).dropna()

            if len(original_values) < 2 or len(synthetic_values) < 2:
                continue

            ks_result = ks_2samp(
                original_values,
                synthetic_values
            )

            similarity = max(
                0, 1 - float(ks_result.statistic)
            )

            numerical_scores.append(similarity)

            results.append({
                "Evaluation": f"Numerical distribution - {col}",
                "Original Value": f"Mean: {original_values.mean():.4f}",
                "Synthetic Value": f"Mean: {synthetic_values.mean():.4f}",
                "Result": f"KS similarity: {similarity * 100:.2f}%"
            })

    numerical_score = (
        np.mean(numerical_scores)
        if numerical_scores else None
    )

    # --------------------------------------------------------
    # 7. CATEGORICAL DISTRIBUTION COMPARISON
    # --------------------------------------------------------

    categorical_scores = []

    for col in common_columns:

        is_original_numeric = pd.api.types.is_numeric_dtype(
            original[col]
        )

        is_synthetic_numeric = pd.api.types.is_numeric_dtype(
            synthetic[col]
        )

        if is_original_numeric or is_synthetic_numeric:
            continue

        original_values = original[col].dropna().astype(str)
        synthetic_values = synthetic[col].dropna().astype(str)

        if len(original_values) == 0 or len(synthetic_values) == 0:
            continue

        original_dist = original_values.value_counts(normalize=True)
        synthetic_dist = synthetic_values.value_counts(normalize=True)

        all_categories = original_dist.index.union(
            synthetic_dist.index
        )

        original_dist = original_dist.reindex(
            all_categories, fill_value=0
        )

        synthetic_dist = synthetic_dist.reindex(
            all_categories, fill_value=0
        )

        tvd = 0.5 * np.abs(
            original_dist - synthetic_dist
        ).sum()

        similarity = max(0, 1 - float(tvd))

        categorical_scores.append(similarity)

        results.append({
            "Evaluation": f"Categorical distribution - {col}",
            "Original Value": f"{original_values.nunique()} unique values",
            "Synthetic Value": f"{synthetic_values.nunique()} unique values",
            "Result": f"Similarity: {similarity * 100:.2f}%"
        })

    categorical_score = (
        np.mean(categorical_scores)
        if categorical_scores else None
    )

    # --------------------------------------------------------
    # 8. OVERALL QUALITY SCORE
    # --------------------------------------------------------

    score_components = [
        column_match,
        type_score,
        missing_score,
        duplicate_score
    ]

    if numerical_score is not None:
        score_components.append(numerical_score)

    if categorical_score is not None:
        score_components.append(categorical_score)

    overall_score = (
        float(np.mean(score_components)) * 100
    )

    # --------------------------------------------------------
    # 9. FINAL EVALUATION REPORT
    # --------------------------------------------------------

    if overall_score >= 85:
        quality_label = "Very Good"
    elif overall_score >= 70:
        quality_label = "Good"
    elif overall_score >= 50:
        quality_label = "Moderate"
    else:
        quality_label = "Needs Improvement"

    report.append("")
    report.append("## Quality Evaluation Results")
    report.append(
        f"### Overall Synthetic Data Quality Score: **{overall_score:.2f}/100**"
    )
    report.append(f"### Quality Assessment: **{quality_label}**")

    report.append("")
    report.append("### Individual Evaluation Scores")
    report.append(f"- Column consistency: **{column_match * 100:.2f}%**")
    report.append(f"- Data type consistency: **{type_score * 100:.2f}%**")
    report.append(f"- Missing-value similarity: **{missing_score * 100:.2f}%**")
    report.append(f"- Duplicate-rate similarity: **{duplicate_score * 100:.2f}%**")

    if numerical_score is not None:
        report.append(
            f"- Numerical distribution similarity: **{numerical_score * 100:.2f}%**"
        )

    if categorical_score is not None:
        report.append(
            f"- Categorical distribution similarity: **{categorical_score * 100:.2f}%**"
        )

    report.append("")
    report.append("### Evaluation Summary")
    report.append(
        "The synthetic dataset was compared with the original dataset "
        "using structural checks, missing-value analysis, duplicate-rate "
        "comparison, and statistical distribution similarity."
    )
    report.append(
        "A higher score indicates closer similarity across the evaluated "
        "checks. This score is an internal project metric, not a formal "
        "privacy or statistical certification."
    )

    results_df = pd.DataFrame(results)

    return "\n".join(report), results_df


# ============================================================
# 10. GRADIO INTERFACE
# ============================================================

with gr.Blocks(theme=gr.themes.Soft()) as quality_demo:

    gr.Markdown("""
    # SynData Arena — Synthetic Data Quality Evaluation

    Evaluate the quality of generated synthetic data by comparing
    it with the original uploaded dataset.
    """)

    evaluate_button = gr.Button(
        "Evaluate Synthetic Data Quality",
        variant="primary"
    )

    quality_report = gr.Markdown(
        "Click the button to start quality evaluation."
    )

    quality_table = gr.Dataframe(
        label="Detailed Quality Evaluation Results",
        interactive=False
    )

    evaluate_button.click(
        fn=evaluate_synthetic_quality,
        inputs=[],
        outputs=[
            quality_report,
            quality_table
        ]
    )


# ============================================================
# CELL 10 INITIALIZATION
# ============================================================

print("================================================")
print("   SYNTHETIC DATA QUALITY EVALUATION READY")
print("================================================")
print("Column Consistency Check: READY")
print("Data Type Validation: READY")
print("Missing Value Comparison: READY")
print("Duplicate Record Analysis: READY")
print("Numerical Distribution Evaluation: READY")
print("Categorical Distribution Evaluation: READY")
print("Overall Quality Score: READY")
print("Gradio Quality Evaluation Interface: READY")
print("================================================")
```

```python

# ============================================================
# CELL 11: AUTOMATIC REGRESSION + CLASSIFICATION ML TESTING
# ============================================================

import pandas as pd
import numpy as np
import gradio as gr
import warnings
import plotly.graph_objects as go

from sklearn.base import clone
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

from sklearn.ensemble import (
    RandomForestRegressor,
    RandomForestClassifier
)

from sklearn.linear_model import (
    LinearRegression,
    LogisticRegression
)

from sklearn.tree import (
    DecisionTreeRegressor,
    DecisionTreeClassifier
)

from sklearn.neighbors import (
    KNeighborsRegressor,
    KNeighborsClassifier
)

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

warnings.filterwarnings("ignore")


# ============================================================
# 1. TARGET COLUMN DROPDOWN
# ============================================================

def get_target_columns():

    global uploaded_df

    if uploaded_df is None or uploaded_df.empty:
        return []

    return [str(col) for col in uploaded_df.columns]


def refresh_ml_columns():

    columns = get_target_columns()

    if not columns:
        return (
            gr.Dropdown(
                choices=[],
                value=None,
                label="Select Target Column"
            ),
            "Please upload a CSV dataset first."
        )

    return (
        gr.Dropdown(
            choices=columns,
            value=columns[0],
            label="Select Target Column"
        ),
        f"{len(columns)} columns available. Select any target column."
    )


# ============================================================
# 2. IMAGE DATASET DETECTION
# ============================================================

def ml_is_image_dataset():

    status = globals().get(
        "image_extraction_status", {}
    )

    if not isinstance(status, dict):
        return False

    processing = str(
        status.get("processing", "")
    ).upper()

    description = str(
        status.get("description", "")
    ).upper()

    return (
        "BLIP" in processing
        or "BLIP" in description
    )


# ============================================================
# 3. AUTOMATIC REGRESSION OR CLASSIFICATION DETECTION
# ============================================================

def detect_ml_problem(target):

    numeric_target = pd.to_numeric(
        target,
        errors="coerce"
    )

    numeric_ratio = numeric_target.notna().mean()

    if numeric_ratio < 0.90:
        return "Classification", target.astype(str)

    numeric_target = numeric_target.astype(float)

    unique_count = numeric_target.nunique()
    total_count = len(numeric_target)

    if total_count == 0:
        return "Classification", target.astype(str)

    threshold = min(
        20,
        max(2, int(np.sqrt(total_count)))
    )

    if (
        unique_count <= threshold
        and unique_count / total_count <= 0.10
    ):
        return "Classification", numeric_target

    return "Regression", numeric_target


# ============================================================
# 4. CREATE FOUR ML MODELS
# ============================================================

def get_ml_models(problem_type, training_rows):

    if problem_type == "Regression":

        return {
            "Random Forest": RandomForestRegressor(
                n_estimators=150,
                random_state=42,
                n_jobs=-1
            ),

            "Linear Regression": LinearRegression(),

            "Decision Tree": DecisionTreeRegressor(
                random_state=42
            ),

            "KNN": KNeighborsRegressor(
                n_neighbors=min(5, training_rows)
            )
        }

    return {
        "Random Forest": RandomForestClassifier(
            n_estimators=150,
            random_state=42,
            n_jobs=-1
        ),

        "Logistic Regression": LogisticRegression(
            max_iter=2000,
            random_state=42
        ),

        "Decision Tree": DecisionTreeClassifier(
            random_state=42
        ),

        "KNN": KNeighborsClassifier(
            n_neighbors=min(5, training_rows)
        )
    }


# ============================================================
# 5. PREPROCESSING
# ============================================================

def create_ml_preprocessor(X_train):

    numerical_columns = X_train.select_dtypes(
        include=np.number
    ).columns.tolist()

    categorical_columns = [
        col for col in X_train.columns
        if col not in numerical_columns
    ]

    transformers = []

    if numerical_columns:

        numerical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            )
        ])

        transformers.append((
            "numerical",
            numerical_pipeline,
            numerical_columns
        ))

    if categorical_columns:

        categorical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="most_frequent")
            ),
            (
                "encoder",
                OneHotEncoder(handle_unknown="ignore")
            )
        ])

        transformers.append((
            "categorical",
            categorical_pipeline,
            categorical_columns
        ))

    return ColumnTransformer(
        transformers=transformers,
        remainder="drop"
    )


# ============================================================
# 6. MAIN ML TESTING FUNCTION
# ============================================================

def _run_ml_testing_core(target_column):

    global uploaded_df

    empty_table = pd.DataFrame(columns=[
        "Model",
        "R2 Score",
        "MAE",
        "RMSE",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "Status"
    ])

    # --------------------------------------------------------
    # VALIDATE DATASET
    # --------------------------------------------------------

    if uploaded_df is None or uploaded_df.empty:
        return empty_table, None, (
            "Please upload a CSV dataset first."
        )

    if ml_is_image_dataset():
        return empty_table, None, (
            "Machine Learning Model Testing is available "
            "for CSV datasets only. Please upload a CSV dataset."
        )

    if not target_column:
        return empty_table, None, (
            "Please select a target column."
        )

    df = uploaded_df.copy()

    if target_column not in df.columns.astype(str).tolist():
        return empty_table, None, (
            "Target column not found. Please refresh the dropdown."
        )

    actual_target = next(
        col for col in df.columns
        if str(col) == target_column
    )

    try:

        df = df.replace(
            [np.inf, -np.inf],
            np.nan
        )

        # ----------------------------------------------------
        # DETECT PROBLEM TYPE
        # ----------------------------------------------------

        problem_type, y = detect_ml_problem(
            df[actual_target]
        )

        valid_rows = y.notna()

        X = df.drop(
            columns=[actual_target]
        ).loc[valid_rows].copy()

        y = y.loc[valid_rows].copy()

        X = X.dropna(
            axis=1,
            how="all"
        )

        if len(X) < 10:
            return empty_table, None, (
                "Not enough valid records. Please use a dataset "
                "with at least 10 valid rows."
            )

        if y.nunique() < 2:
            return empty_table, None, (
                "The selected target has only one unique value. "
                "Please select another target column."
            )

        # ----------------------------------------------------
        # REMOVE ID-LIKE FEATURE COLUMNS
        # ----------------------------------------------------

        id_columns = []

        for col in X.columns:

            name = str(col).lower().strip()

            if (
                name == "id"
                or name.endswith("_id")
                or name.endswith("id")
            ):
                id_columns.append(col)

        for col in X.columns:

            if col in id_columns:
                continue

            if (
                X[col].nunique(dropna=True)
                / max(len(X), 1)
            ) > 0.95:
                id_columns.append(col)

        X = X.drop(
            columns=id_columns,
            errors="ignore"
        )

        if X.shape[1] == 0:
            return empty_table, None, (
                "No usable feature columns remain after "
                "removing ID columns."
            )

        # ----------------------------------------------------
        # CONVERT DATETIME FEATURES
        # ----------------------------------------------------

        for col in X.columns:

            if pd.api.types.is_datetime64_any_dtype(X[col]):
                X[col] = X[col].astype(str)

        # ----------------------------------------------------
        # TRAIN-TEST SPLIT
        # ----------------------------------------------------

        stratify_values = None

        if problem_type == "Classification":

            class_counts = y.value_counts()

            if class_counts.min() < 2:
                return empty_table, None, (
                    "Classification requires at least two records "
                    "for every target class. Please select another "
                    "target or provide more records."
                )

            if y.nunique() > 50:
                return empty_table, None, (
                    "This target has too many distinct classes "
                    "for reliable automatic classification. "
                    "Please choose another target."
                )

            stratify_values = y

        try:

            X_train, X_test, y_train, y_test = train_test_split(
                X,
                y,
                test_size=0.20,
                random_state=42,
                stratify=stratify_values
            )

        except Exception as split_error:

            return empty_table, None, (
                f"Could not split the dataset: {str(split_error)}"
            )

        if len(X_train) < 1 or len(X_test) < 1:
            return empty_table, None, (
                "Insufficient training or testing records."
            )

        # ----------------------------------------------------
        # CREATE SYNTHETIC TRAINING DATA
        # ----------------------------------------------------

        rng = np.random.default_rng(42)

        sample_positions = rng.integers(
            0,
            len(X_train),
            size=len(X_train)
        )

        X_synthetic_train = (
            X_train.iloc[sample_positions]
            .copy()
            .reset_index(drop=True)
        )

        y_synthetic_train = (
            y_train.iloc[sample_positions]
            .copy()
            .reset_index(drop=True)
        )

        # Add controlled numerical variations
        numeric_features = X_synthetic_train.select_dtypes(
            include=np.number
        ).columns

        for col in numeric_features:

            original_values = pd.to_numeric(
                X_train[col],
                errors="coerce"
            ).dropna()

            if len(original_values) < 2:
                continue

            std_value = original_values.std()

            if pd.isna(std_value) or std_value == 0:
                continue

            noise = rng.normal(
                loc=0,
                scale=std_value * 0.02,
                size=len(X_synthetic_train)
            )

            generated_values = pd.to_numeric(
                X_synthetic_train[col],
                errors="coerce"
            )

            X_synthetic_train[col] = (
                generated_values + noise
            ).clip(
                lower=original_values.min(),
                upper=original_values.max()
            )

        # ----------------------------------------------------
        # CREATE FOUR APPROPRIATE MODELS
        # ----------------------------------------------------

        models = get_ml_models(
            problem_type,
            len(X_synthetic_train)
        )

        preprocessor = create_ml_preprocessor(
            X_train
        )

        results = []

        # ----------------------------------------------------
        # TRAIN AND EVALUATE ALL MODELS
        # ----------------------------------------------------

        for model_name, model in models.items():

            try:

                pipeline = Pipeline([
                    (
                        "preprocessor",
                        clone(preprocessor)
                    ),
                    (
                        "model",
                        clone(model)
                    )
                ])

                pipeline.fit(
                    X_synthetic_train,
                    y_synthetic_train
                )

                predictions = pipeline.predict(X_test)

                if problem_type == "Regression":

                    r2 = r2_score(
                        y_test,
                        predictions
                    )

                    mae = mean_absolute_error(
                        y_test,
                        predictions
                    )

                    rmse = np.sqrt(
                        mean_squared_error(
                            y_test,
                            predictions
                        )
                    )

                    results.append({
                        "Model": model_name,
                        "R2 Score": round(float(r2), 4),
                        "MAE": round(float(mae), 4),
                        "RMSE": round(float(rmse), 4),
                        "Accuracy": "N/A",
                        "Precision": "N/A",
                        "Recall": "N/A",
                        "F1 Score": "N/A",
                        "Status": "Successful"
                    })

                else:

                    accuracy = accuracy_score(
                        y_test,
                        predictions
                    )

                    precision = precision_score(
                        y_test,
                        predictions,
                        average="weighted",
                        zero_division=0
                    )

                    recall = recall_score(
                        y_test,
                        predictions,
                        average="weighted",
                        zero_division=0
                    )

                    f1 = f1_score(
                        y_test,
                        predictions,
                        average="weighted",
                        zero_division=0
                    )

                    results.append({
                        "Model": model_name,
                        "R2 Score": "N/A",
                        "MAE": "N/A",
                        "RMSE": "N/A",
                        "Accuracy": round(float(accuracy), 4),
                        "Precision": round(float(precision), 4),
                        "Recall": round(float(recall), 4),
                        "F1 Score": round(float(f1), 4),
                        "Status": "Successful"
                    })

            except Exception as model_error:

                results.append({
                    "Model": model_name,
                    "R2 Score": "N/A",
                    "MAE": "N/A",
                    "RMSE": "N/A",
                    "Accuracy": "N/A",
                    "Precision": "N/A",
                    "Recall": "N/A",
                    "F1 Score": "N/A",
                    "Status": (
                        f"Failed: {str(model_error)[:120]}"
                    )
                })

        results_df = pd.DataFrame(results)

        successful = results_df[
            results_df["Status"] == "Successful"
        ].copy()

        if successful.empty:
            return (
                results_df,
                None,
                "All four models failed. Please check your dataset."
            )

        # ----------------------------------------------------
        # SELECT BEST MODEL
        # ----------------------------------------------------

        if problem_type == "Regression":

            successful["_score"] = pd.to_numeric(
                successful["R2 Score"]
            )

            successful["_error"] = pd.to_numeric(
                successful["RMSE"]
            )

            successful = successful.sort_values(
                by=["_score", "_error"],
                ascending=[False, True]
            )

            graph_metric = "R2 Score"

        else:

            successful["_score"] = pd.to_numeric(
                successful["F1 Score"]
            )

            successful["_error"] = 1 - pd.to_numeric(
                successful["Accuracy"]
            )

            successful = successful.sort_values(
                by=["_score", "_error"],
                ascending=[False, True]
            )

            graph_metric = "F1 Score"

        best = successful.iloc[0]
        best_name = best["Model"]

        # ----------------------------------------------------
        # CREATE ONE MODEL COMPARISON GRAPH
        # ----------------------------------------------------

        graph_data = results_df[
            results_df["Status"] == "Successful"
        ].copy()

        graph_values = pd.to_numeric(
            graph_data[graph_metric]
        )

        graph_colors = [
            "#16A34A" if name == best_name else "#64748B"
            for name in graph_data["Model"]
        ]

        fig = go.Figure()

        fig.add_trace(go.Bar(
            x=graph_data["Model"],
            y=graph_values,
            marker_color=graph_colors,
            text=graph_values.round(4).astype(str),
            textposition="outside"
        ))

        fig.update_layout(
            title=(
                f"Four ML Models Comparison — {graph_metric}"
            ),
            xaxis_title="Machine Learning Models",
            yaxis_title=graph_metric,
            template="plotly_white",
            showlegend=False,
            height=430,
            margin=dict(t=70, b=60)
        )

        # ----------------------------------------------------
        # BEST MODEL EXPLANATION
        # ----------------------------------------------------

        report = [
            "# ML Model Testing Results",
            "",
            f"**Selected Target:** {target_column}",
            f"**Automatically Detected Problem:** {problem_type}",
            f"**Models Successfully Tested:** {len(successful)} out of 4",
            "",
            "# Best Performing Model",
            "",
            f"## {best_name}",
            ""
        ]

        if problem_type == "Regression":

            report.extend([
                f"**R² Score:** {best['R2 Score']}",
                f"**MAE:** {best['MAE']}",
                f"**RMSE:** {best['RMSE']}",
                "",
                f"The {best_name} achieved the highest R² score among the tested models.",
                f"Its RMSE was {best['RMSE']}, indicating its prediction error on the original test records.",
                "It is the best-performing model in this experiment."
            ])

        else:

            report.extend([
                f"**Accuracy:** {best['Accuracy']}",
                f"**Precision:** {best['Precision']}",
                f"**Recall:** {best['Recall']}",
                f"**F1 Score:** {best['F1 Score']}",
                "",
                f"The {best_name} achieved the highest weighted F1 Score, providing the best balance between precision and recall.",
                f"Its accuracy was {best['Accuracy']}.",
                "It is the best-performing model in this experiment."
            ])

        report.extend([
            "",
            "# Why the Other Models Performed Less Well"
        ])

        for _, row in successful.iterrows():

            if row["Model"] == best_name:
                continue

            if problem_type == "Regression":

                report.append(
                    f"- **{row['Model']}:** R² = {row['R2 Score']}, "
                    f"RMSE = {row['RMSE']}. Its test-set R² "
                    f"was lower than the best-performing model."
                )

            else:

                report.append(
                    f"- **{row['Model']}:** Accuracy = {row['Accuracy']}, "
                    f"F1 Score = {row['F1 Score']}. Its weighted "
                    f"F1 Score was lower than the best-performing model."
                )

        failed_models = results_df[
            results_df["Status"] != "Successful"
        ]

        for _, row in failed_models.iterrows():

            report.append(
                f"- **{row['Model']}:** This model could not be evaluated. "
                f"Reason: {row['Status']}"
            )

        report.extend([
            "",
            "# Testing Details",
            f"- Valid records: {len(X):,}",
            f"- Training records: {len(X_train):,}",
            f"- Testing records: {len(X_test):,}",
            f"- Input features used: {X.shape[1]}",
            "- Training data: Resampled synthetic training records",
            "- Testing data: Original held-out records",
            "",
            "*Note: Synthetic training records are generated using resampling and small numerical variations. This is an experimental testing approach, not a formal privacy-guaranteed synthetic data generation method.*"
        ])

        final_report = "\n".join(report)

        return results_df, fig, final_report

    except Exception as e:

        traceback.print_exc()

        return (
            empty_table,
            None,
            f"ML testing encountered an error: {str(e)}"
        )


# ============================================================
# 7. PDF REPORT COMPATIBILITY
# ============================================================

def run_ml_testing(target_column):

    results_df, _, report = _run_ml_testing_core(
        target_column
    )

    return report, results_df


# ============================================================
# 8. GRADIO INTERFACE
# ============================================================

initial_columns = get_target_columns()

with gr.Blocks(
    theme=gr.themes.Soft()
) as ml_testing_demo:

    gr.Markdown("""
    # SynData Arena — Machine Learning Model Testing

    Select any target column. The platform automatically detects
    whether the task is Regression or Classification and selects
    the appropriate four ML algorithms and evaluation metrics.
    """)

    with gr.Row():

        target_dropdown = gr.Dropdown(
            choices=initial_columns,
            value=(
                initial_columns[0]
                if initial_columns else None
            ),
            label="Select Target Column",
            interactive=True
        )

        refresh_button = gr.Button(
            "Refresh Columns"
        )

    column_status = gr.Markdown(
        "Select the target column you want to predict."
        if initial_columns
        else "Upload a CSV dataset, then click Refresh Columns."
    )

    run_ml_button = gr.Button(
        "Run ML Model Testing",
        variant="primary"
    )

    ml_results_table = gr.Dataframe(
        label="Four ML Models — Performance Comparison",
        interactive=False
    )

    ml_comparison_plot = gr.Plot(
        label="ML Model Comparison Graph"
    )

    ml_report = gr.Markdown(
        "Your best model and explanation will appear here."
    )

    refresh_button.click(
        fn=refresh_ml_columns,
        inputs=[],
        outputs=[
            target_dropdown,
            column_status
        ]
    )

    run_ml_button.click(
        fn=_run_ml_testing_core,
        inputs=[target_dropdown],
        outputs=[
            ml_results_table,
            ml_comparison_plot,
            ml_report
        ]
    )


# ============================================================
# 9. CELL INITIALIZATION
# ============================================================

print("================================================")
print("       AUTOMATIC ML TESTING READY")
print("================================================")
print("Regression Detection: READY")
print("Classification Detection: READY")
print("Random Forest: READY")
print("Linear / Logistic Regression: READY")
print("Decision Tree: READY")
print("KNN: READY")
print("Performance Comparison Table: READY")
print("One Comparison Graph: READY")
print("Best Model Identification: READY")
print("Model Explanation: READY")
print("PDF Compatibility Function: READY")
print("================================================")
```

```python

# ============================================================
# CELL 12: SYNDATA ARENA - ONE-CLICK COMPLETE PROJECT PDF
# UPDATED: SAFE GENERATION + NO ML RERUN
# ============================================================

import os
import re
import io
import pandas as pd
import numpy as np
import gradio as gr
import matplotlib.pyplot as plt

from datetime import datetime

from reportlab.lib import colors
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.units import mm
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Table,
    TableStyle,
    PageBreak,
    Image as PDFImage
)
from xml.sax.saxutils import escape


# ============================================================
# 1. PDF STYLES
# ============================================================

def create_pdf_styles():

    base = getSampleStyleSheet()

    return {
        "title": ParagraphStyle(
            "PDFTitle",
            parent=base["Title"],
            fontName="Helvetica-Bold",
            fontSize=25,
            leading=30,
            alignment=TA_CENTER,
            textColor=colors.HexColor("#243B64"),
            spaceAfter=14
        ),

        "subtitle": ParagraphStyle(
            "PDFSubtitle",
            parent=base["Normal"],
            fontSize=10,
            leading=15,
            alignment=TA_CENTER,
            textColor=colors.grey,
            spaceAfter=12
        ),

        "heading": ParagraphStyle(
            "PDFHeading",
            parent=base["Heading2"],
            fontName="Helvetica-Bold",
            fontSize=15,
            leading=19,
            textColor=colors.HexColor("#243B64"),
            spaceBefore=12,
            spaceAfter=8
        ),

        "subheading": ParagraphStyle(
            "PDFSubheading",
            parent=base["Heading3"],
            fontSize=10,
            leading=14,
            textColor=colors.HexColor("#355C8A"),
            spaceBefore=8,
            spaceAfter=5
        ),

        "normal": ParagraphStyle(
            "PDFNormal",
            parent=base["BodyText"],
            fontSize=8.5,
            leading=13,
            spaceAfter=6
        ),

        "small": ParagraphStyle(
            "PDFSmall",
            parent=base["BodyText"],
            fontSize=7,
            leading=10,
            textColor=colors.grey
        ),

        "table": ParagraphStyle(
            "PDFTable",
            parent=base["BodyText"],
            fontSize=6.5,
            leading=9,
            wordWrap="CJK"
        ),

        "table_header": ParagraphStyle(
            "PDFTableHeader",
            parent=base["BodyText"],
            fontName="Helvetica-Bold",
            fontSize=7,
            leading=9,
            textColor=colors.white,
            wordWrap="CJK"
        )
    }


# ============================================================
# 2. TEXT HELPER
# ============================================================

def pdf_add_text(story, value, style):

    if value is None:
        return

    text = str(value)

    text = escape(text)
    text = re.sub(r"\*\*(.*?)\*\*", r"<b>\1</b>", text)
    text = re.sub(r"\*(.*?)\*", r"<i>\1</i>", text)
    text = text.replace("\n", "<br/>")

    story.append(Paragraph(text, style))
    story.append(Spacer(1, 5))


def pdf_add_heading(story, heading, styles, level="heading"):

    story.append(Spacer(1, 8))
    pdf_add_text(story, heading, styles[level])


# ============================================================
# 3. TABLE HELPER
# ============================================================

def pdf_add_table(story, dataframe, styles):

    if dataframe is None:

        pdf_add_text(
            story,
            "Results have not been generated yet.",
            styles["normal"]
        )
        return

    if not isinstance(dataframe, pd.DataFrame):

        try:
            dataframe = pd.DataFrame(dataframe)

        except Exception:

            pdf_add_text(
                story,
                str(dataframe),
                styles["normal"]
            )
            return

    if dataframe.empty:

        pdf_add_text(
            story,
            "No results available for this section.",
            styles["normal"]
        )
        return

    display_df = dataframe.head(40).copy()

    table_data = [[
        Paragraph(
            escape(str(col)),
            styles["table_header"]
        )
        for col in display_df.columns
    ]]

    for _, row in display_df.iterrows():

        table_data.append([
            Paragraph(
                escape(str(value))[:300],
                styles["table"]
            )
            for value in row.tolist()
        ])

    available_width = A4[0] - 30 * mm
    column_count = max(1, len(display_df.columns))

    table = Table(
        table_data,
        colWidths=[
            available_width / column_count
        ] * column_count,
        repeatRows=1,
        hAlign="LEFT"
    )

    table.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0),
         colors.HexColor("#243B64")),
        ("GRID", (0, 0), (-1, -1), 0.4, colors.grey),
        ("VALIGN", (0, 0), (-1, -1), "TOP"),
        ("ROWBACKGROUNDS", (0, 1), (-1, -1),
         [colors.white, colors.HexColor("#F1F4F8")]),
        ("LEFTPADDING", (0, 0), (-1, -1), 4),
        ("RIGHTPADDING", (0, 0), (-1, -1), 4),
        ("TOPPADDING", (0, 0), (-1, -1), 5),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 5)
    ]))

    story.append(table)
    story.append(Spacer(1, 10))

    if len(dataframe) > 40:

        pdf_add_text(
            story,
            "Only the first 40 rows are included in this table.",
            styles["small"]
        )


# ============================================================
# 4. GRAPH HELPER
# ============================================================

def pdf_add_figure(story, figure, explanation, styles):

    if figure is not None:

        try:

            image_buffer = io.BytesIO()

            if hasattr(figure, "savefig"):

                figure.savefig(
                    image_buffer,
                    format="png",
                    dpi=140,
                    bbox_inches="tight"
                )

            elif hasattr(figure, "to_image"):

                image_buffer.write(
                    figure.to_image(format="png")
                )

            else:
                raise TypeError("Unsupported graph format")

            image_buffer.seek(0)

            story.append(
                PDFImage(
                    image_buffer,
                    width=165 * mm,
                    height=88 * mm,
                    kind="proportional"
                )
            )

            story.append(Spacer(1, 5))

        except Exception as error:

            pdf_add_text(
                story,
                f"Graph could not be embedded: {str(error)[:150]}",
                styles["small"]
            )

    if explanation:

        pdf_add_text(
            story,
            f"Graph Explanation: {explanation}",
            styles["normal"]
        )


# ============================================================
# 5. DATASET OVERVIEW
# ============================================================

def create_pdf_dataset_overview(df):

    return pd.DataFrame({
        "Property": [
            "Dataset Name",
            "Total Records",
            "Total Columns",
            "Duplicate Records",
            "Total Missing Values",
            "Numerical Columns",
            "Categorical Columns"
        ],

        "Value": [
            globals().get(
                "uploaded_csv_name",
                None
            ) or globals().get(
                "uploaded_image_name",
                None
            ) or "Uploaded Dataset",

            len(df),
            len(df.columns),
            int(df.duplicated().sum()),
            int(df.isnull().sum().sum()),

            len(df.select_dtypes(
                include=np.number
            ).columns),

            len(df.select_dtypes(
                include=["object", "category", "bool"]
            ).columns)
        ]
    })


# ============================================================
# 6. GET EXISTING ML RESULTS
# IMPORTANT: DOES NOT RUN ML AGAIN
# ============================================================

def get_saved_ml_results():

    possible_tables = [
        "ml_results_df",
        "ml_results",
        "saved_ml_results",
        "latest_ml_results"
    ]

    possible_reports = [
        "ml_report_text",
        "saved_ml_report",
        "latest_ml_report"
    ]

    possible_graphs = [
        "ml_comparison_fig",
        "saved_ml_graph",
        "latest_ml_graph"
    ]

    result_table = None
    result_report = None
    result_graph = None

    for name in possible_tables:

        value = globals().get(name)

        if isinstance(value, pd.DataFrame) and not value.empty:
            result_table = value.copy()
            break

    for name in possible_reports:

        value = globals().get(name)

        if value:
            result_report = str(value)
            break

    for name in possible_graphs:

        value = globals().get(name)

        if value is not None:
            result_graph = value
            break

    return result_table, result_report, result_graph


# ============================================================
# 7. COMPLETE PROJECT PDF GENERATOR
# ============================================================

def generate_project_pdf():

    global uploaded_df
    global synthetic_df
    global synthetic_generation_summary

    if uploaded_df is None or uploaded_df.empty:

        raise gr.Error(
            "Please upload a CSV dataset or image dataset first."
        )

    try:

        styles = create_pdf_styles()
        story = []

        original_df = uploaded_df.copy()

        dataset_overview = create_pdf_dataset_overview(
            original_df
        )

        # ----------------------------------------------------
        # ORIGINAL DATASET ANALYSIS
        # ----------------------------------------------------

        analysis_report = "Dataset analysis has not been generated yet."
        quality_report = "Data quality analysis has not been generated yet."
        quality_breakdown = None
        ai_insights = "Dataset insights have not been generated yet."
        graph_items = []

        try:

            if callable(globals().get("run_dataset_analysis")):

                analysis_output = run_dataset_analysis()

                if isinstance(analysis_output, (tuple, list)):

                    if len(analysis_output) >= 12:

                        analysis_report = analysis_output[0]
                        quality_report = analysis_output[1]
                        quality_breakdown = analysis_output[2]
                        ai_insights = analysis_output[3]

                        graph_items = [
                            (analysis_output[4], analysis_output[5]),
                            (analysis_output[6], analysis_output[7]),
                            (analysis_output[8], analysis_output[9]),
                            (analysis_output[10], analysis_output[11])
                        ]

                    elif len(analysis_output) >= 8:

                        analysis_report = analysis_output[0]
                        quality_report = analysis_output[1]
                        quality_breakdown = analysis_output[2]
                        ai_insights = analysis_output[3]

                        graph_items = [
                            (analysis_output[4], analysis_output[5]),
                            (analysis_output[6], analysis_output[7])
                        ]

        except Exception as analysis_error:

            analysis_report = (
                "Dataset analysis could not be collected: "
                + str(analysis_error)
            )

        # ----------------------------------------------------
        # SYNTHETIC DATA
        # ----------------------------------------------------

        current_synthetic = globals().get("synthetic_df")
        generation_summary = globals().get(
            "synthetic_generation_summary"
        )

        has_synthetic = (
            isinstance(current_synthetic, pd.DataFrame)
            and not current_synthetic.empty
        )

        # ----------------------------------------------------
        # SYNTHETIC QUALITY
        # ----------------------------------------------------

        synthetic_quality_report = (
            "Synthetic data quality evaluation has not been completed."
        )

        synthetic_quality_table = None

        if has_synthetic:

            try:

                synthetic_quality_report, synthetic_quality_table = (
                    evaluate_synthetic_quality()
                )

            except Exception as quality_error:

                synthetic_quality_report = (
                    "Quality evaluation could not be collected: "
                    + str(quality_error)
                )

        # ----------------------------------------------------
        # ML RESULTS - USE SAVED RESULTS ONLY
        # ----------------------------------------------------

        ml_results, ml_report, ml_graph = get_saved_ml_results()

        if ml_results is None:

            ml_results = pd.DataFrame()

        if not ml_report:

            ml_report = (
                "ML testing results have not been saved for this report. "
                "Run ML Model Testing after updating Cell 11 to store "
                "its results."
            )

        # ----------------------------------------------------
        # PDF FILE
        # ----------------------------------------------------

        output_path = os.path.join(
            "/tmp",
            "SynData_Arena_Complete_Project_Report.pdf"
        )

        document = SimpleDocTemplate(
            output_path,
            pagesize=A4,
            rightMargin=15 * mm,
            leftMargin=15 * mm,
            topMargin=15 * mm,
            bottomMargin=15 * mm,
            title="SynData Arena Complete Project Report",
            author="SynData Arena"
        )

        # ----------------------------------------------------
        # COVER PAGE
        # ----------------------------------------------------

        story.append(Spacer(1, 35 * mm))

        story.append(
            Paragraph(
                "SYNDATA ARENA",
                styles["title"]
            )
        )

        story.append(
            Paragraph(
                "Synthetic Data Generation and ML Testing Platform",
                styles["subtitle"]
            )
        )

        story.append(Spacer(1, 20))

        pdf_add_heading(
            story,
            "COMPLETE PROJECT REPORT",
            styles
        )

        pdf_add_text(
            story,
            "This report presents the results available in the "
            "SynData Arena platform, including dataset analysis, "
            "data quality assessment, synthetic data generation, "
            "quality evaluation and machine learning testing.",
            styles["normal"]
        )

        pdf_add_text(
            story,
            f"Report Generated: {datetime.now().strftime('%d-%m-%Y %H:%M:%S')}",
            styles["normal"]
        )

        pdf_add_table(
            story,
            dataset_overview,
            styles
        )

        if has_synthetic:

            pdf_add_text(
                story,
                f"Synthetic Records Generated: {len(current_synthetic):,}",
                styles["normal"]
            )

        else:

            pdf_add_text(
                story,
                "Synthetic Data: Not generated yet.",
                styles["normal"]
            )

        story.append(PageBreak())

        # ----------------------------------------------------
        # SECTION 1: ORIGINAL DATASET ANALYSIS
        # ----------------------------------------------------

        pdf_add_heading(
            story,
            "1. ORIGINAL DATASET ANALYSIS",
            styles
        )

        pdf_add_text(
            story,
            str(analysis_report),
            styles["normal"]
        )

        pdf_add_heading(
            story,
            "Dataset Overview",
            styles,
            "subheading"
        )

        pdf_add_table(
            story,
            dataset_overview,
            styles
        )

        # ----------------------------------------------------
        # SECTION 2: DATA QUALITY
        # ----------------------------------------------------

        pdf_add_heading(
            story,
            "2. DATA QUALITY ASSESSMENT",
            styles
        )

        pdf_add_text(
            story,
            str(quality_report),
            styles["normal"]
        )

        pdf_add_table(
            story,
            quality_breakdown,
            styles
        )

        # ----------------------------------------------------
        # SECTION 3: AI INSIGHTS
        # ----------------------------------------------------

        pdf_add_heading(
            story,
            "3. DATASET INSIGHTS",
            styles
        )

        pdf_add_text(
            story,
            str(ai_insights),
            styles["normal"]
        )

        # ----------------------------------------------------
        # SECTION 4: VISUALIZATIONS
        # ----------------------------------------------------

        pdf_add_heading(
            story,
            "4. DATASET VISUALIZATIONS",
            styles
        )

        if graph_items:

            for index, (figure, explanation) in enumerate(
                graph_items,
                start=1
            ):

                pdf_add_heading(
                    story,
                    f"Visualization {index}",
                    styles,
                    "subheading"
                )

                pdf_add_figure(
                    story,
                    figure,
                    explanation,
                    styles
                )

        else:

            pdf_add_text(
                story,
                "Dataset visualization figures are not available yet.",
                styles["normal"]
            )

        # ----------------------------------------------------
        # SECTION 5: SYNTHETIC DATA GENERATION
        # ----------------------------------------------------

        pdf_add_heading(
            story,
            "5. SYNTHETIC DATA GENERATION",
            styles
        )

        if has_synthetic:

            pdf_add_text(
                story,
                f"Original Records: {len(original_df):,}",
                styles["normal"]
            )

            pdf_add_text(
                story,
                f"Synthetic Records: {len(current_synthetic):,}",
                styles["normal"]
            )

            pdf_add_text(
                story,
                f"Number of Columns: {len(current_synthetic.columns)}",
                styles["normal"]
            )

            pdf_add_heading(
                story,
                "Synthetic Generation Summary",
                styles,
                "subheading"
            )

            pdf_add_text(
                story,
                str(generation_summary or "Summary not available."),
                styles["normal"]
            )

            pdf_add_heading(
                story,
                "Synthetic Dataset Preview",
                styles,
                "subheading"
            )

            pdf_add_table(
                story,
                current_synthetic.head(10),
                styles
            )

        else:

            pdf_add_text(
                story,
                "Synthetic data generation has not been completed yet.",
                styles["normal"]
            )

        # ----------------------------------------------------
        # SECTION 6: SYNTHETIC QUALITY
        # ----------------------------------------------------

        pdf_add_heading(
            story,
            "6. SYNTHETIC DATA QUALITY EVALUATION",
            styles
        )

        pdf_add_text(
            story,
            str(synthetic_quality_report),
            styles["normal"]
        )

        pdf_add_table(
            story,
            synthetic_quality_table,
            styles
        )

        # ----------------------------------------------------
        # SECTION 7: ML TESTING
        # ----------------------------------------------------

        pdf_add_heading(
            story,
            "7. MACHINE LEARNING MODEL TESTING",
            styles
        )

        pdf_add_text(
            story,
            str(ml_report),
            styles["normal"]
        )

        pdf_add_heading(
            story,
            "Model Performance Comparison",
            styles,
            "subheading"
        )

        pdf_add_table(
            story,
            ml_results,
            styles
        )

        if ml_graph is not None:

            pdf_add_figure(
                story,
                ml_graph,
                "Machine learning model comparison.",
                styles
            )

        # ----------------------------------------------------
        # SECTION 8: PROJECT CONCLUSION
        # ----------------------------------------------------

        pdf_add_heading(
            story,
            "8. PROJECT CONCLUSION",
            styles
        )

        conclusion = (
            "SynData Arena integrates dataset analysis, data quality "
            "assessment, dataset insights, synthetic data generation, "
            "synthetic data quality evaluation and machine learning "
            "testing into a unified platform. The platform helps users "
            "understand datasets, generate synthetic records and "
            "evaluate their usefulness through machine learning."
        )

        pdf_add_text(
            story,
            conclusion,
            styles["normal"]
        )

        pdf_add_text(
            story,
            "Important Note: Quality scores are internal project "
            "evaluation measures and do not certify privacy protection. "
            "Machine learning performance depends on the dataset, "
            "selected target and experimental settings.",
            styles["small"]
        )

        # ----------------------------------------------------
        # BUILD AND VERIFY PDF
        # ----------------------------------------------------

        document.build(story)

        if not os.path.exists(output_path):
            raise ValueError("PDF file could not be created.")

        if os.path.getsize(output_path) == 0:
            raise ValueError("Generated PDF is empty.")

        print("Complete SynData Arena PDF generated successfully.")
        print("File:", output_path)
        print("File size:", os.path.getsize(output_path), "bytes")

        return output_path

    except Exception as error:

        print("PDF generation error:", str(error))

        raise gr.Error(
            f"PDF generation failed: {str(error)}"
        )


# ============================================================
# 8. ONE-BUTTON PDF INTERFACE
# ============================================================

with gr.Blocks(theme=gr.themes.Soft()) as pdf_report_demo:

    gr.Markdown("""
    # Complete Project PDF Report

    Download one complete report containing the results
    available from the SynData Arena modules.

    You can download the report even if some modules
    have not been completed yet.
    """)

    pdf_download_button = gr.Button(
        "Download Complete Project PDF",
        variant="primary"
    )

    pdf_download = gr.File(
        label="Your Complete SynData Arena PDF",
        interactive=False
    )

    pdf_download_button.click(
        fn=generate_project_pdf,
        inputs=[],
        outputs=[pdf_download]
    )


# ============================================================
# 9. INITIALIZATION
# ============================================================

print("================================================")
print("       COMPLETE PROJECT PDF READY")
print("================================================")
print("Original Dataset Analysis: INCLUDED")
print("Data Quality Assessment: INCLUDED")
print("Dataset Insights: INCLUDED")
print("Dataset Visualizations: INCLUDED")
print("Synthetic Data Summary: OPTIONAL")
print("Synthetic Quality Evaluation: OPTIONAL")
print("ML Results: SAVED RESULTS ONLY")
print("ML Testing Re-run During PDF: DISABLED")
print("One-Button PDF Download: READY")
print("================================================")
```

```python

# ============================================================
# CELL 13: ML PERFORMANCE COMPARISON AND VISUALIZATION
# UPDATED: REGRESSION + CLASSIFICATION + PDF RESULT SAVING
# ============================================================

import pandas as pd
import numpy as np
import gradio as gr
import plotly.express as px


# ============================================================
# 1. ML PERFORMANCE COMPARISON FUNCTION
# ============================================================

def generate_ml_comparison(target_column):

    global uploaded_df
    global ml_results_df
    global ml_report_text
    global ml_comparison_fig
    global ml_comparison_figures

    if uploaded_df is None or uploaded_df.empty:

        return (
            "Please upload a dataset first.",
            pd.DataFrame(),
            None,
            None,
            None
        )

    if not target_column:

        return (
            "Please select a target column.",
            pd.DataFrame(),
            None,
            None,
            None
        )

    try:

        # Run ML testing once when the user clicks this button.
        report, results_df = run_ml_testing(target_column)

        if results_df is None or results_df.empty:

            return (
                report,
                pd.DataFrame(),
                None,
                None,
                None
            )

        comparison_df = results_df.copy()

        if "Status" in comparison_df.columns:

            comparison_df = comparison_df[
                comparison_df["Status"].astype(str).str.lower()
                == "successful"
            ].copy()

        if comparison_df.empty:

            return (
                "No models completed successfully. Please check Cell 11.",
                results_df,
                None,
                None,
                None
            )

        # ----------------------------------------------------
        # DETECT REGRESSION OR CLASSIFICATION
        # ----------------------------------------------------

        classification_metrics = [
            "Accuracy",
            "Precision",
            "Recall",
            "F1 Score"
        ]

        regression_metrics = [
            "R2 Score",
            "MAE",
            "RMSE"
        ]

        available_classification = []
        available_regression = []

        for metric in classification_metrics + regression_metrics:

            if metric in comparison_df.columns:

                comparison_df[metric] = pd.to_numeric(
                    comparison_df[metric],
                    errors="coerce"
                )

                if comparison_df[metric].notna().any():

                    if metric in classification_metrics:
                        available_classification.append(metric)

                    else:
                        available_regression.append(metric)

        if available_classification:

            problem_type = "Classification"
            metrics = available_classification

            primary_metric = (
                "F1 Score"
                if "F1 Score" in metrics
                else metrics[0]
            )

            higher_is_better = True

        elif available_regression:

            problem_type = "Regression"
            metrics = available_regression

            primary_metric = (
                "R2 Score"
                if "R2 Score" in metrics
                else metrics[0]
            )

            higher_is_better = primary_metric == "R2 Score"

        else:

            return (
                "No supported performance metrics were found. "
                "Please check the results generated by Cell 11.",
                results_df,
                None,
                None,
                None
            )

        comparison_df = comparison_df.dropna(
            subset=[primary_metric]
        ).copy()

        if comparison_df.empty:

            return (
                "No valid model performance values are available.",
                results_df,
                None,
                None,
                None
            )

        # ----------------------------------------------------
        # CREATE THREE METRIC GRAPHS
        # ----------------------------------------------------

        figures = []

        selected_metrics = metrics[:3]

        for metric in selected_metrics:

            plot_df = comparison_df.dropna(
                subset=[metric]
            ).copy()

            if plot_df.empty:
                figures.append(None)
                continue

            color_column = (
                "Training Dataset"
                if "Training Dataset" in plot_df.columns
                else None
            )

            fig = px.bar(
                plot_df,
                x="Model",
                y=metric,
                color=color_column,
                barmode="group",
                title=f"{metric} Comparison",
                text_auto=".3f",
                labels={
                    "Model": "Machine Learning Model",
                    metric: metric,
                    "Training Dataset": "Training Data"
                }
            )

            fig.update_layout(
                xaxis_title="Machine Learning Models",
                yaxis_title=metric,
                template="plotly_white",
                height=420
            )

            if metric in ["MAE", "RMSE"]:
                fig.update_layout(
                    yaxis_title=f"{metric} (Lower is Better)"
                )

            else:
                fig.update_layout(
                    yaxis_title=f"{metric} (Higher is Better)"
                )

            figures.append(fig)

        while len(figures) < 3:
            figures.append(None)

        # ----------------------------------------------------
        # IDENTIFY BEST MODEL
        # ----------------------------------------------------

        synthetic_results = comparison_df.copy()

        if "Training Dataset" in synthetic_results.columns:

            synthetic_mask = (
                synthetic_results["Training Dataset"]
                .astype(str)
                .str.contains("synthetic", case=False, na=False)
            )

            if synthetic_mask.any():

                synthetic_results = synthetic_results[
                    synthetic_mask
                ].copy()

        ascending = not higher_is_better

        best_model = synthetic_results.sort_values(
            by=primary_metric,
            ascending=ascending
        ).iloc[0]

        best_name = best_model["Model"]
        best_score = float(best_model[primary_metric])

        best_explanation = (
            f"## Best Performing Model: {best_name}\n\n"
            f"The best-performing model is **{best_name}** "
            f"with a {primary_metric} of **{best_score:.4f}**.\n\n"
        )

        if problem_type == "Classification":

            best_explanation += (
                "Classification evaluates how accurately the model "
                "predicts categories. F1 Score balances precision "
                "and recall, making it a useful comparison metric."
            )

        else:

            best_explanation += (
                "Regression evaluates predictions of numerical values. "
                "R² measures explained variation, while MAE and RMSE "
                "measure prediction errors."
            )

        # ----------------------------------------------------
        # FINAL COMPARISON REPORT
        # ----------------------------------------------------

        metric_explanations = ""

        if problem_type == "Classification":

            metric_explanations = (
                "- **Accuracy:** Overall correct predictions.\n"
                "- **Precision:** Correct positive predictions "
                "among predicted positives.\n"
                "- **Recall:** Correct positive predictions "
                "among actual positives.\n"
                "- **F1 Score:** Balance between precision and recall."
            )

        else:

            metric_explanations = (
                "- **R²:** Higher is generally better.\n"
                "- **MAE:** Lower is better.\n"
                "- **RMSE:** Lower is better and penalizes "
                "larger prediction errors."
            )

        final_report = (
            f"# ML PERFORMANCE COMPARISON\n\n"
            f"**Target Column:** {target_column}\n\n"
            f"**Problem Type:** {problem_type}\n\n"
            f"**Models Compared:** Random Forest, Linear / Logistic "
            f"Regression, Decision Tree and KNN.\n\n"
            f"**Evaluation:** Models are evaluated using the "
            f"held-out original test dataset.\n\n"
            f"{best_explanation}\n\n"
            f"### Understanding the Metrics\n"
            f"{metric_explanations}\n\n"
            f"**Note:** The comparison uses the model results "
            f"generated by Cell 11."
        )

        # ----------------------------------------------------
        # SAVE RESULTS FOR CELL 12 PDF
        # ----------------------------------------------------

        ml_results_df = comparison_df.copy()
        ml_report_text = final_report
        ml_comparison_figures = figures
        ml_comparison_fig = next(
            (fig for fig in figures if fig is not None),
            None
        )

        print("ML comparison results saved for PDF.")
        print("Problem type:", problem_type)
        print("Best model:", best_name)

        return (
            final_report,
            comparison_df,
            figures[0],
            figures[1],
            figures[2]
        )

    except Exception as error:

        print("ML comparison error:", str(error))

        return (
            f"ML comparison failed: {str(error)}",
            pd.DataFrame(),
            None,
            None,
            None
        )


# ============================================================
# 2. GRADIO INTERFACE
# ============================================================

comparison_columns = get_target_columns()

with gr.Blocks(theme=gr.themes.Soft()) as ml_comparison_demo:

    gr.Markdown("""
    # SynData Arena — ML Performance Comparison

    Compare machine learning models trained using
    original and synthetic training data.
    """)

    with gr.Row():

        comparison_target_dropdown = gr.Dropdown(
            choices=comparison_columns,
            value=(
                comparison_columns[0]
                if comparison_columns else None
            ),
            label="Select Target Column",
            interactive=True
        )

        comparison_refresh_button = gr.Button(
            "Refresh Columns"
        )

    comparison_status = gr.Markdown(
        "Select a target column to compare model performance."
    )

    comparison_run_button = gr.Button(
        "Generate Performance Comparison",
        variant="primary"
    )

    comparison_report = gr.Markdown(
        "Your comparison results will appear here."
    )

    comparison_table = gr.Dataframe(
        label="Model Performance Comparison Table",
        interactive=False
    )

    gr.Markdown("## Performance Visualization")

    r2_output = gr.Plot(
        label="Performance Metric 1"
    )

    mae_output = gr.Plot(
        label="Performance Metric 2"
    )

    rmse_output = gr.Plot(
        label="Performance Metric 3"
    )

    comparison_refresh_button.click(
        fn=refresh_ml_columns,
        inputs=[],
        outputs=[
            comparison_target_dropdown,
            comparison_status
        ]
    )

    comparison_run_button.click(
        fn=generate_ml_comparison,
        inputs=[comparison_target_dropdown],
        outputs=[
            comparison_report,
            comparison_table,
            r2_output,
            mae_output,
            rmse_output
        ]
    )


# ============================================================
# 3. INITIALIZATION
# ============================================================

print("================================================")
print("   ML PERFORMANCE COMPARISON READY")
print("================================================")
print("Regression Metrics: READY")
print("Classification Metrics: READY")
print("Original vs Synthetic Comparison: READY")
print("Performance Comparison Table: READY")
print("Best Model Identification: READY")
print("Three Metric Visualizations: READY")
print("PDF Result Saving: READY")
print("================================================")
```

```python

# ============================================================
# FIXED SYNTHETIC DATA GENERATION ENGINE
# Fix: assignment destination is read-only
# ============================================================

import pandas as pd
import numpy as np
import os
import traceback

# Global variables
synthetic_df = None
synthetic_file_path = None
synthetic_generation_summary = None


def generate_synthetic_data(number_of_records):

    global uploaded_df
    global synthetic_df
    global synthetic_file_path
    global synthetic_generation_summary

    try:

        # ----------------------------------------------------
        # 1. VALIDATE INPUT
        # ----------------------------------------------------

        if uploaded_df is None:
            return (
                pd.DataFrame(),
                "ERROR: Please upload a CSV dataset first.",
                None
            )

        if not isinstance(uploaded_df, pd.DataFrame):
            return (
                pd.DataFrame(),
                "ERROR: Uploaded dataset is invalid.",
                None
            )

        if uploaded_df.empty:
            return (
                pd.DataFrame(),
                "ERROR: Uploaded dataset is empty.",
                None
            )

        try:
            number_of_records = int(number_of_records)
        except (ValueError, TypeError):
            return (
                pd.DataFrame(),
                "ERROR: Enter a valid record count.",
                None
            )

        if not 1 <= number_of_records <= 10000:
            return (
                pd.DataFrame(),
                "ERROR: Record count must be between 1 and 10,000.",
                None
            )

        # ----------------------------------------------------
        # 2. CREATE INDEPENDENT WRITABLE DATASET COPY
        # ----------------------------------------------------

        original_df = uploaded_df.copy(deep=True)

        rng = np.random.default_rng()

        original_rows = len(original_df)

        # Bootstrap sampling
        selected_indices = rng.integers(
            low=0,
            high=original_rows,
            size=number_of_records
        )

        synthetic = (
            original_df.iloc[selected_indices]
            .copy(deep=True)
            .reset_index(drop=True)
        )

        # ----------------------------------------------------
        # 3. IDENTIFY COLUMN TYPES
        # ----------------------------------------------------

        numeric_columns = original_df.select_dtypes(
            include=[np.number]
        ).columns.tolist()

        categorical_columns = [
            col for col in original_df.columns
            if col not in numeric_columns
        ]

        generated_numeric = 0
        generated_categorical = 0
        generated_identifiers = 0

        # ----------------------------------------------------
        # 4. GENERATE NUMERICAL VALUES
        # ----------------------------------------------------

        for col in numeric_columns:

            try:

                original_values = pd.to_numeric(
                    original_df[col],
                    errors="coerce"
                ).to_numpy(dtype=float, copy=True)

                new_values = pd.to_numeric(
                    synthetic[col],
                    errors="coerce"
                ).to_numpy(dtype=float, copy=True)

                valid_original = original_values[
                    np.isfinite(original_values)
                ]

                if len(valid_original) == 0:
                    synthetic[col] = new_values
                    continue

                unique_count = len(np.unique(valid_original))

                # Perturb continuous columns only
                if unique_count > 2:

                    std_value = float(np.std(valid_original))

                    if np.isfinite(std_value) and std_value > 0:

                        noise = rng.normal(
                            loc=0,
                            scale=std_value * 0.03,
                            size=number_of_records
                        )

                        valid_new = np.isfinite(new_values)

                        new_values[valid_new] = (
                            new_values[valid_new] +
                            noise[valid_new]
                        )

                        minimum = float(np.min(valid_original))
                        maximum = float(np.max(valid_original))

                        new_values[valid_new] = np.clip(
                            new_values[valid_new],
                            minimum,
                            maximum
                        )

                # Preserve integer columns
                if pd.api.types.is_integer_dtype(original_df[col].dtype):
                    valid_new = np.isfinite(new_values)
                    new_values[valid_new] = np.rint(
                        new_values[valid_new]
                    )

                # Assign a new writable array
                synthetic[col] = new_values

                generated_numeric += 1

            except Exception as column_error:

                print(
                    f"Numerical column skipped ({col}): "
                    f"{column_error}"
                )

        # ----------------------------------------------------
        # 5. HANDLE CATEGORICAL COLUMNS
        # ----------------------------------------------------

        for col in categorical_columns:

            try:

                # Bootstrap already preserves sampled categories.
                # Create an independent copy for safe assignment.
                values = synthetic[col].to_numpy(copy=True)

                synthetic[col] = pd.Series(
                    values,
                    index=synthetic.index
                )

                generated_categorical += 1

            except Exception as column_error:

                print(
                    f"Categorical column skipped ({col}): "
                    f"{column_error}"
                )

        # ----------------------------------------------------
        # 6. GENERATE UNIQUE IDENTIFIERS
        # ----------------------------------------------------

        for col in original_df.columns:

            col_name = str(col).strip().lower()

            is_identifier = (
                col_name == "id"
                or col_name.endswith("_id")
                or col_name.endswith("id")
            )

            if not is_identifier:
                continue

            try:

                if pd.api.types.is_numeric_dtype(original_df[col].dtype):

                    valid_ids = pd.to_numeric(
                        original_df[col],
                        errors="coerce"
                    ).dropna()

                    start_id = (
                        int(valid_ids.max()) + 1
                        if len(valid_ids) else 1
                    )

                    new_ids = np.arange(
                        start_id,
                        start_id + number_of_records,
                        dtype=np.int64
                    )

                    synthetic[col] = new_ids

                else:

                    new_ids = [
                        f"SYN_{i:06d}"
                        for i in range(1, number_of_records + 1)
                    ]

                    synthetic[col] = new_ids

                generated_identifiers += 1

            except Exception as id_error:

                print(
                    f"Identifier generation skipped ({col}): "
                    f"{id_error}"
                )

        # ----------------------------------------------------
        # 7. PRESERVE ORIGINAL COLUMN ORDER
        # ----------------------------------------------------

        synthetic = synthetic[original_df.columns].copy()

        # ----------------------------------------------------
        # 8. SAVE SYNTHETIC CSV
        # ----------------------------------------------------

        output_path = "/tmp/SynData_Arena_Synthetic_Data.csv"

        synthetic.to_csv(
            output_path,
            index=False,
            encoding="utf-8-sig"
        )

        # ----------------------------------------------------
        # 9. SAVE GLOBAL RESULTS
        # ----------------------------------------------------

        synthetic_df = synthetic.copy(deep=True)

        synthetic_file_path = output_path

        synthetic_generation_summary = {
            "original_records": original_rows,
            "synthetic_records": number_of_records,
            "total_columns": len(synthetic.columns),
            "numerical_columns": generated_numeric,
            "categorical_columns": generated_categorical,
            "identifier_columns": generated_identifiers
        }

        # ----------------------------------------------------
        # 10. GENERATION SUMMARY
        # ----------------------------------------------------

        summary = f"""
## Synthetic Data Generation Successful!

| Metric | Result |
|---|---:|
| Original Records | {original_rows:,} |
| Generated Records | {number_of_records:,} |
| Total Columns | {len(synthetic.columns)} |
| Numerical Columns | {generated_numeric} |
| Categorical Columns | {generated_categorical} |
| Identifier Columns | {generated_identifiers} |

**Generation Status:** Successful

**Output File:** SynData_Arena_Synthetic_Data.csv

Your synthetic dataset is ready to download.
"""

        return (
            synthetic,
            summary,
            output_path
        )

    except Exception as e:

        error_message = (
            "ERROR DURING SYNTHETIC DATA GENERATION: "
            + str(e)
        )

        print(traceback.format_exc())

        return (
            pd.DataFrame(),
            error_message,
            None
        )


print("================================================")
print(" FIXED SYNTHETIC DATA GENERATION ENGINE READY")
print("================================================")
print("Writable DataFrame Copies: READY")
print("Numerical Data Generation: READY")
print("Categorical Data Handling: READY")
print("Unique Identifier Generation: READY")
print("Missing Value Handling: READY")
print("CSV Export: READY")
print("Record Limit: 1 to 10,000")
print("================================================")
```

```python

# ============================================================
# CELL 15: FINAL SYNDATA ARENA APPLICATION
# 7 TABS | CSV + BLIP IMAGE SUPPORT
# COMPLETE PDF REPORT | PUBLIC GRADIO URL
# ============================================================

import gradio as gr
import pandas as pd
import numpy as np
import os
import tempfile
import html
import traceback
import matplotlib.pyplot as plt

from reportlab.lib import colors
from reportlab.lib.pagesizes import A4
from reportlab.lib.units import inch
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, Table,
    TableStyle, Image as PDFImage, PageBreak
)

print("Building SynData Arena...")


# ============================================================
# 1. PRESERVE CELL 14 HANDLERS
# ============================================================

cell14_analysis_handler = globals().get("final_run_dataset_analysis")
cell14_image_detector = globals().get("final_is_image_dataset")


# ============================================================
# 2. RESULTS STORAGE
# ============================================================

syndata_results = {
    "analysis": None,
    "synthetic": None,
    "quality": None,
    "ml_report": None,
    "ml_table": None,
    "ml_graph": None
}


def clear_previous_results():

    for key in syndata_results:
        syndata_results[key] = None

    globals()["ml_results_df"] = None
    globals()["ml_report_text"] = None
    globals()["ml_comparison_fig"] = None


# ============================================================
# 3. DATASET HELPERS
# ============================================================

def get_current_dataset():

    for name in [
        "uploaded_df",
        "original_df",
        "df",
        "dataset",
        "data"
    ]:

        value = globals().get(name)

        if isinstance(value, pd.DataFrame) and not value.empty:
            return value.copy()

    return None


def check_dataset():

    current_df = get_current_dataset()

    return current_df is not None and not current_df.empty


def is_image_dataset():

    current_df = get_current_dataset()

    if current_df is None:
        return False

    if callable(cell14_image_detector):
        return cell14_image_detector(current_df)

    status = globals().get("image_extraction_status", {})

    if isinstance(status, dict):

        if status.get("processing") in [
            "BLIP Local AI",
            "BLIP Local Vision"
        ]:
            return True

    if "source" in current_df.columns:

        return current_df["source"].astype(str).str.contains(
            "BLIP Local AI|BLIP Local Vision",
            case=False,
            na=False
        ).any()

    return False


def get_dataset_columns():

    current_df = get_current_dataset()

    if current_df is None:
        return []

    try:

        if callable(globals().get("get_target_columns")):
            return list(get_target_columns())

    except Exception:
        pass

    return list(current_df.columns)


# ============================================================
# 4. CSV UPLOAD
# ============================================================

def final_upload_csv(file_path):

    try:

        if file_path is None:

            return (
                "Please upload a CSV file.",
                pd.DataFrame(),
                gr.update(choices=[], value=None)
            )

        status, preview = upload_csv(file_path)

        current_df = get_current_dataset()

        if current_df is None:

            return (
                status,
                pd.DataFrame(),
                gr.update(choices=[], value=None)
            )

        clear_previous_results()

        columns = get_dataset_columns()

        return (
            status,
            current_df.head(10),
            gr.update(
                choices=columns,
                value=columns[0] if columns else None
            )
        )

    except Exception as e:

        print(traceback.format_exc())

        return (
            f"CSV upload failed: {str(e)}",
            pd.DataFrame(),
            gr.update(choices=[], value=None)
        )


# ============================================================
# 5. IMAGE UPLOAD
# CELL 3 BLIP LOCAL VISION - NO GEMINI
# ============================================================

def final_upload_image(image):

    try:

        if image is None:

            return (
                "Please upload an image.",
                pd.DataFrame(),
                gr.update(choices=[], value=None)
            )

        status, preview = upload_image(image)

        current_df = get_current_dataset()

        if current_df is None or current_df.empty:

            return (
                status,
                pd.DataFrame(),
                gr.update(choices=[], value=None)
            )

        globals()["uploaded_df"] = current_df.copy()

        clear_previous_results()

        columns = get_dataset_columns()

        return (
            status,
            current_df.head(10),
            gr.update(
                choices=columns,
                value=columns[0] if columns else None
            )
        )

    except Exception as e:

        print(traceback.format_exc())

        return (
            f"Image upload failed: {str(e)}",
            pd.DataFrame(),
            gr.update(choices=[], value=None)
        )


# ============================================================
# 6. DATASET ANALYSIS
# ============================================================

def final_run_dataset_analysis():

    try:

        if not check_dataset():

            return (
                "Please upload a CSV or image first.",
                "",
                pd.DataFrame(),
                "",
                None, "",
                None, "",
                None, "",
                None, ""
            )

        if not callable(cell14_analysis_handler):
            raise Exception(
                "Cell 14 analysis handler is unavailable. Run Cell 14 first."
            )

        result = cell14_analysis_handler()

        if result and len(result) >= 12:
            syndata_results["analysis"] = result

        return result

    except Exception as e:

        print(traceback.format_exc())

        return (
            f"Dataset analysis failed: {str(e)}",
            "",
            pd.DataFrame(),
            "",
            None, "",
            None, "",
            None, "",
            None, ""
        )


# ============================================================
# 7. SYNTHETIC DATA GENERATION
# ============================================================

def final_generate_synthetic_data(record_count):

    try:

        if not check_dataset():

            return (
                pd.DataFrame(),
                "Please upload a dataset first.",
                None
            )

        if is_image_dataset():

            return (
                pd.DataFrame(),
                "Synthetic statistical data generation is available "
                "for CSV datasets. A single image-derived record is "
                "not suitable for this operation.",
                None
            )

        result = generate_synthetic_data(record_count)

        if result and len(result) >= 3:

            syndata_results["synthetic"] = {
                "preview": result[0],
                "summary": result[1],
                "file": result[2]
            }

        return result

    except Exception as e:

        print(traceback.format_exc())

        return (
            pd.DataFrame(),
            f"Synthetic generation failed: {str(e)}",
            None
        )


# ============================================================
# 8. QUALITY EVALUATION
# ============================================================

def final_evaluate_synthetic_quality():

    try:

        if not check_dataset():

            return (
                "Please upload a dataset first.",
                pd.DataFrame()
            )

        if is_image_dataset():

            return (
                "Statistical synthetic data quality evaluation "
                "is available for CSV datasets. A single image "
                "does not provide enough observations.",
                pd.DataFrame()
            )

        result = evaluate_synthetic_quality()

        if result and len(result) >= 2:
            syndata_results["quality"] = result

        return result

    except Exception as e:

        print(traceback.format_exc())

        return (
            f"Quality evaluation failed: {str(e)}",
            pd.DataFrame()
        )


# ============================================================
# 9. ML MODEL TESTING
# CELL 11 CORE RETURNS:
# RESULTS DATAFRAME, GRAPH, REPORT
# ============================================================

def final_run_ml_testing(target_column):

    try:

        if not check_dataset():

            return (
                "Please upload a CSV dataset first.",
                pd.DataFrame()
            )

        if is_image_dataset():

            return (
                "ML testing requires multiple labelled observations. "
                "Please upload a CSV dataset for ML model testing.",
                pd.DataFrame()
            )

        if not target_column:

            return (
                "Please select a target column.",
                pd.DataFrame()
            )

        results_df, fig, report = _run_ml_testing_core(
            target_column
        )

        syndata_results["ml_report"] = report
        syndata_results["ml_table"] = results_df
        syndata_results["ml_graph"] = fig

        # Also save for compatibility with the PDF/Cell 13 code.
        globals()["ml_results_df"] = results_df
        globals()["ml_report_text"] = report
        globals()["ml_comparison_fig"] = fig

        return report, results_df

    except Exception as e:

        print(traceback.format_exc())

        return (
            f"ML testing failed: {str(e)}",
            pd.DataFrame()
        )


# ============================================================
# 10. SYNBOT
# ============================================================

def final_safe_synbot(message, history):

    try:

        if not message or not str(message).strip():
            return "Please enter a question."

        if not check_dataset():
            return "Please upload a CSV or image dataset first."

        return synbot_ai(message)

    except Exception as e:

        print(traceback.format_exc())

        return f"SynBot error: {str(e)}"


# ============================================================
# 11. REFRESH ML TARGET COLUMNS
# ============================================================

def final_refresh_ml_columns():

    try:

        if not check_dataset():

            return (
                gr.update(choices=[], value=None),
                "Please upload a dataset first."
            )

        columns = get_dataset_columns()

        return (
            gr.update(
                choices=columns,
                value=columns[0] if columns else None
            ),
            f"Found {len(columns)} columns."
        )

    except Exception as e:

        return (
            gr.update(choices=[], value=None),
            str(e)
        )


# ============================================================
# 12. PDF HELPER FUNCTIONS
# ============================================================

def pdf_text(value):

    if value is None:
        return "Not available"

    return html.escape(str(value)).replace("\n", "<br/>")


def pdf_heading(story, heading, styles):

    story.append(Spacer(1, 12))
    story.append(
        Paragraph(pdf_text(heading), styles["Heading2"])
    )
    story.append(Spacer(1, 6))


def pdf_paragraph(story, content, styles):

    story.append(
        Paragraph(pdf_text(content), styles["BodyText"])
    )
    story.append(Spacer(1, 8))


def pdf_dataframe(
    story,
    dataframe,
    styles,
    max_rows=25,
    max_columns=8
):

    if not isinstance(dataframe, pd.DataFrame):

        try:
            dataframe = pd.DataFrame(dataframe)

        except Exception:
            pdf_paragraph(story, "No table available.", styles)
            return

    if dataframe.empty:

        pdf_paragraph(story, "No results available.", styles)
        return

    display_df = dataframe.iloc[:max_rows, :max_columns]

    table_data = [[
        Paragraph(
            "<b>" + pdf_text(column) + "</b>",
            styles["BodyText"]
        )
        for column in display_df.columns
    ]]

    for _, row in display_df.iterrows():

        table_data.append([
            Paragraph(pdf_text(value), styles["BodyText"])
            for value in row.tolist()
        ])

    available_width = A4[0] - 1.1 * inch
    column_width = available_width / len(display_df.columns)

    table = Table(
        table_data,
        colWidths=[column_width] * len(display_df.columns),
        repeatRows=1
    )

    table.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#695CFF")),
        ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
        ("GRID", (0, 0), (-1, -1), 0.4, colors.grey),
        ("VALIGN", (0, 0), (-1, -1), "TOP"),
        ("LEFTPADDING", (0, 0), (-1, -1), 4),
        ("RIGHTPADDING", (0, 0), (-1, -1), 4),
        ("TOPPADDING", (0, 0), (-1, -1), 4),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 4)
    ]))

    story.append(table)
    story.append(Spacer(1, 10))


def pdf_graph(
    story,
    figure,
    explanation,
    folder,
    filename,
    styles
):

    if figure is not None:

        try:

            image_path = os.path.join(folder, filename)

            # Matplotlib figure
            if hasattr(figure, "savefig"):

                figure.savefig(
                    image_path,
                    dpi=150,
                    bbox_inches="tight"
                )

            # Plotly figure
            elif hasattr(figure, "write_image"):

                figure.write_image(image_path)

            else:
                raise Exception("Unsupported graph format.")

            if os.path.exists(image_path):

                story.append(
                    PDFImage(
                        image_path,
                        width=6.2 * inch,
                        height=3.8 * inch,
                        kind="proportional"
                    )
                )

                story.append(Spacer(1, 6))

        except Exception as e:

            pdf_paragraph(
                story,
                f"Graph could not be embedded: {str(e)}",
                styles
            )

    if explanation:
        pdf_paragraph(story, explanation, styles)


# ============================================================
# 13. COMPLETE PDF REPORT
# NO ML RERUN
# WORKS EVEN IF SOME MODULES HAVE NOT BEEN RUN
# ============================================================

def generate_final_project_pdf():

    try:

        original_df = get_current_dataset()

        if original_df is None:

            return (
                "Please upload a dataset before generating the report.",
                None
            )

        temp_folder = tempfile.mkdtemp(
            prefix="syndata_report_"
        )

        pdf_path = os.path.join(
            temp_folder,
            "SynData_Arena_Complete_Report.pdf"
        )

        styles = getSampleStyleSheet()

        title_style = ParagraphStyle(
            "SynDataTitle",
            parent=styles["Title"],
            fontSize=22,
            leading=28,
            alignment=TA_CENTER,
            textColor=colors.HexColor("#695CFF")
        )

        story = []

        # TITLE PAGE
        story.append(Spacer(1, 65))
        story.append(
            Paragraph("SYNDATA ARENA", title_style)
        )
        story.append(Spacer(1, 15))

        pdf_paragraph(
            story,
            "AI-Powered Synthetic Data Generation and ML Testing Platform",
            styles
        )

        story.append(Spacer(1, 20))

        story.append(
            Paragraph(
                "Complete Project Report",
                styles["Heading1"]
            )
        )

        pdf_paragraph(
            story,
            f"Dataset records: {len(original_df)}",
            styles
        )

        pdf_paragraph(
            story,
            f"Dataset columns: {len(original_df.columns)}",
            styles
        )

        story.append(PageBreak())

        # SECTION 1: DATASET
        pdf_heading(
            story,
            "1. Original Dataset Overview",
            styles
        )

        pdf_paragraph(
            story,
            f"Total records: {len(original_df)}",
            styles
        )

        pdf_paragraph(
            story,
            f"Total columns: {len(original_df.columns)}",
            styles
        )

        pdf_paragraph(
            story,
            "Columns: " + ", ".join(map(str, original_df.columns)),
            styles
        )

        pdf_heading(
            story,
            "Dataset Preview",
            styles
        )

        pdf_dataframe(
            story,
            original_df.head(10),
            styles,
            max_rows=10
        )

        # SECTION 2: ANALYSIS
        pdf_heading(
            story,
            "2. Dataset Analysis",
            styles
        )

        analysis = syndata_results["analysis"]

        if analysis:

            pdf_heading(story, "Dataset Overview", styles)
            pdf_paragraph(story, analysis[0], styles)

            pdf_heading(story, "Data Quality", styles)
            pdf_paragraph(story, analysis[1], styles)

            pdf_dataframe(story, analysis[2], styles)

            pdf_heading(story, "AI Insights", styles)
            pdf_paragraph(story, analysis[3], styles)

            graph_items = [
                (4, 5, "Dataset Graph 1", "graph1.png"),
                (6, 7, "Dataset Graph 2", "graph2.png"),
                (8, 9, "Dataset Graph 3", "graph3.png"),
                (10, 11, "Dataset Graph 4", "graph4.png")
            ]

            for fig_index, exp_index, heading, filename in graph_items:

                pdf_heading(story, heading, styles)

                pdf_graph(
                    story,
                    analysis[fig_index],
                    analysis[exp_index],
                    temp_folder,
                    filename,
                    styles
                )

        else:

            pdf_paragraph(
                story,
                "Dataset analysis has not been run yet.",
                styles
            )

        # SECTION 3: SYNTHETIC DATA
        pdf_heading(
            story,
            "3. Synthetic Data Generation",
            styles
        )

        synthetic_info = syndata_results["synthetic"]

        if synthetic_info:

            pdf_paragraph(
                story,
                synthetic_info.get("summary", "Generated successfully."),
                styles
            )

        else:

            pdf_paragraph(
                story,
                "Synthetic data generation has not been run yet.",
                styles
            )

        current_synthetic_df = globals().get("synthetic_df")

        if (
            isinstance(current_synthetic_df, pd.DataFrame)
            and not current_synthetic_df.empty
        ):

            pdf_paragraph(
                story,
                f"Generated records: {len(current_synthetic_df)}",
                styles
            )

            pdf_dataframe(
                story,
                current_synthetic_df.head(10),
                styles,
                max_rows=10
            )

        # SECTION 4: QUALITY
        pdf_heading(
            story,
            "4. Synthetic Data Quality Evaluation",
            styles
        )

        quality = syndata_results["quality"]

        if quality:

            pdf_paragraph(story, quality[0], styles)

            pdf_dataframe(
                story,
                quality[1],
                styles
            )

        else:

            pdf_paragraph(
                story,
                "Quality evaluation has not been run yet.",
                styles
            )

        # SECTION 5: ML
        pdf_heading(
            story,
            "5. Machine Learning Model Testing",
            styles
        )

        if syndata_results["ml_report"] is not None:

            pdf_paragraph(
                story,
                syndata_results["ml_report"],
                styles
            )

            pdf_dataframe(
                story,
                syndata_results["ml_table"],
                styles
            )

            pdf_heading(
                story,
                "Model Performance Comparison Graph",
                styles
            )

            pdf_graph(
                story,
                syndata_results["ml_graph"],
                None,
                temp_folder,
                "ml_comparison.png",
                styles
            )

        else:

            pdf_paragraph(
                story,
                "ML testing has not been run yet.",
                styles
            )

        # SECTION 6: CONCLUSION
        pdf_heading(
            story,
            "6. Project Conclusion",
            styles
        )

        pdf_paragraph(
            story,
            "SynData Arena integrates dataset upload, image understanding, "
            "dataset analysis, synthetic data generation, quality evaluation, "
            "machine learning testing and AI-assisted dataset interaction.",
            styles
        )

        pdf_paragraph(
            story,
            "This report contains the results generated in the application "
            "up to the time of download. Modules that have not been run are "
            "identified accordingly.",
            styles
        )

        # BUILD PDF
        document = SimpleDocTemplate(
            pdf_path,
            pagesize=A4,
            rightMargin=0.55 * inch,
            leftMargin=0.55 * inch,
            topMargin=0.55 * inch,
            bottomMargin=0.55 * inch
        )

        document.build(story)

        if not os.path.exists(pdf_path):
            raise Exception("PDF file was not created.")

        return (
            "Your complete PDF report is ready. Download it below.",
            pdf_path
        )

    except Exception as e:

        print(traceback.format_exc())

        return (
            f"PDF generation failed: {str(e)}",
            None
        )


# ============================================================
# 14. BUILD THE SEVEN-TAB APPLICATION
# ============================================================

initial_columns = get_dataset_columns()

initial_target = (
    initial_columns[0]
    if initial_columns
    else None
)


with gr.Blocks(
    theme=gr.themes.Soft(),
    title="SynData Arena",
    css="""
    .main-header {
        text-align: center;
        padding: 18px;
    }

    .main-header h1 {
        font-size: 36px;
        font-weight: 800;
        color: #695CFF;
    }

    .main-header p {
        font-size: 15px;
    }
    """
) as demo_final:

    gr.HTML("""
    <div class="main-header">
        <h1>SYNDATA ARENA</h1>
        <h3>AI-Powered Synthetic Data Generation and ML Testing Platform</h3>
        <p>
        Upload datasets, generate synthetic data, evaluate quality,
        test machine learning models and download reports.
        </p>
    </div>
    """)

    gr.Markdown("---")

    # --------------------------------------------------------
    # TAB 1: DATASET UPLOAD
    # --------------------------------------------------------

    with gr.Tab("Dataset Upload"):

        gr.Markdown("# Upload Your Dataset")

        with gr.Row():

            with gr.Column():

                gr.Markdown("## CSV Dataset")

                final_csv_upload = gr.File(
                    label="Upload CSV File",
                    file_types=[".csv"],
                    type="filepath"
                )

                final_csv_status = gr.Markdown(
                    "Upload a CSV file to begin."
                )

                final_csv_preview = gr.Dataframe(
                    label="Uploaded Dataset Preview",
                    interactive=False
                )

            with gr.Column():

                gr.Markdown("## Image Upload")

                final_image_upload = gr.Image(
                    label="Upload Image",
                    type="pil"
                )

                final_image_status = gr.Markdown(
                    "Upload a photograph for local BLIP image analysis."
                )

                final_image_preview = gr.Dataframe(
                    label="Extracted Dataset from Image",
                    interactive=False
                )

    # --------------------------------------------------------
    # TAB 2: DATASET ANALYSIS
    # --------------------------------------------------------

    with gr.Tab("Dataset Analysis"):

        gr.Markdown("# Original Dataset Analysis")

        gr.Markdown(
            "Analyze your uploaded CSV or image dataset."
        )

        final_analyze_button = gr.Button(
            "Analyze Dataset",
            variant="primary"
        )

        gr.Markdown("## 1. Dataset Overview")

        final_analysis_report = gr.Markdown()

        gr.Markdown("## 2. Data Quality")

        final_quality_report = gr.Markdown()

        final_analysis_quality_table = gr.Dataframe(
            label="Quality Breakdown",
            interactive=False
        )

        gr.Markdown("## 3. AI Insights")

        final_ai_insights = gr.Markdown()

        gr.Markdown("## 4. Visualizations")

        final_graph1 = gr.Plot()
        final_graph1_explanation = gr.Markdown()

        final_graph2 = gr.Plot()
        final_graph2_explanation = gr.Markdown()

        final_graph3 = gr.Plot()
        final_graph3_explanation = gr.Markdown()

        final_graph4 = gr.Plot()
        final_graph4_explanation = gr.Markdown()

        final_analyze_button.click(
            fn=final_run_dataset_analysis,
            inputs=[],
            outputs=[
                final_analysis_report,
                final_quality_report,
                final_analysis_quality_table,
                final_ai_insights,
                final_graph1,
                final_graph1_explanation,
                final_graph2,
                final_graph2_explanation,
                final_graph3,
                final_graph3_explanation,
                final_graph4,
                final_graph4_explanation
            ]
        )

    # --------------------------------------------------------
    # TAB 3: SYNTHETIC DATA GENERATION
    # --------------------------------------------------------

    with gr.Tab("Synthetic Data Generation"):

        gr.Markdown("# Synthetic Data Generation")

        gr.Markdown(
            "Generate between 1 and 10,000 synthetic records from your CSV dataset."
        )

        final_record_count = gr.Slider(
            minimum=1,
            maximum=10000,
            value=100,
            step=1,
            label="Number of Synthetic Records"
        )

        final_generate_button = gr.Button(
            "Generate Synthetic Data",
            variant="primary"
        )

        final_generation_status = gr.Markdown()

        final_synthetic_preview = gr.Dataframe(
            label="Generated Synthetic Dataset",
            interactive=False
        )

        final_synthetic_download = gr.File(
            label="Download Synthetic CSV"
        )

        final_generate_button.click(
            fn=final_generate_synthetic_data,
            inputs=[final_record_count],
            outputs=[
                final_synthetic_preview,
                final_generation_status,
                final_synthetic_download
            ]
        )

    # --------------------------------------------------------
    # TAB 4: QUALITY EVALUATION
    # --------------------------------------------------------

    with gr.Tab("Quality Evaluation"):

        gr.Markdown("# Synthetic Data Quality Evaluation")

        gr.Markdown(
            "Compare original and synthetic CSV datasets."
        )

        final_evaluation_button = gr.Button(
            "Evaluate Synthetic Data",
            variant="primary"
        )

        final_evaluation_report = gr.Markdown()

        final_evaluation_table = gr.Dataframe(
            label="Detailed Quality Evaluation",
            interactive=False
        )

        final_evaluation_button.click(
            fn=final_evaluate_synthetic_quality,
            inputs=[],
            outputs=[
                final_evaluation_report,
                final_evaluation_table
            ]
        )

    # --------------------------------------------------------
    # TAB 5: ML MODEL TESTING
    # --------------------------------------------------------

    with gr.Tab("ML Model Testing"):

        gr.Markdown("# Machine Learning Model Testing")

        final_ml_dropdown = gr.Dropdown(
            choices=initial_columns,
            value=initial_target,
            label="Select Target Column",
            interactive=True
        )

        final_ml_refresh = gr.Button(
            "Refresh Target Columns"
        )

        final_ml_status = gr.Markdown()

        final_ml_button = gr.Button(
            "Run Machine Learning Testing",
            variant="primary"
        )

        final_ml_report = gr.Markdown()

        final_ml_table = gr.Dataframe(
            label="Machine Learning Model Results",
            interactive=False
        )

        final_ml_refresh.click(
            fn=final_refresh_ml_columns,
            inputs=[],
            outputs=[
                final_ml_dropdown,
                final_ml_status
            ]
        )

        final_ml_button.click(
            fn=final_run_ml_testing,
            inputs=[final_ml_dropdown],
            outputs=[
                final_ml_report,
                final_ml_table
            ]
        )

    # --------------------------------------------------------
    # TAB 6: SYNBOT AI
    # --------------------------------------------------------

    with gr.Tab("SynBot AI"):

        gr.Markdown("# SynBot AI Assistant")

        gr.Markdown(
            "Ask questions about your uploaded dataset."
        )

        gr.ChatInterface(
            fn=final_safe_synbot,
            title="SynBot AI",
            description="Your intelligent dataset assistant."
        )

    # --------------------------------------------------------
    # TAB 7: PDF REPORT
    # --------------------------------------------------------

    with gr.Tab("PDF Report"):

        gr.Markdown("# Complete Project PDF Report")

        gr.Markdown(
            "Download a report containing the results generated so far."
        )

        final_pdf_button = gr.Button(
            "Download PDF Report",
            variant="primary"
        )

        final_pdf_status = gr.Markdown()

        final_pdf_download = gr.File(
            label="Download Complete PDF"
        )

        final_pdf_button.click(
            fn=generate_final_project_pdf,
            inputs=[],
            outputs=[
                final_pdf_status,
                final_pdf_download
            ]
        )

    # --------------------------------------------------------
    # FOOTER
    # --------------------------------------------------------

    gr.Markdown("---")

    gr.HTML("""
    <div style="text-align:center; padding:18px;">
        <h3 style="color:#695CFF;">SynData Arena</h3>
        <p>AI-Powered Synthetic Data Generation and ML Testing Platform</p>
    </div>
    """)

    # --------------------------------------------------------
    # UPLOAD EVENTS
    # --------------------------------------------------------

    final_csv_upload.change(
        fn=final_upload_csv,
        inputs=[final_csv_upload],
        outputs=[
            final_csv_status,
            final_csv_preview,
            final_ml_dropdown
        ]
    )

    final_image_upload.change(
        fn=final_upload_image,
        inputs=[final_image_upload],
        outputs=[
            final_image_status,
            final_image_preview,
            final_ml_dropdown
        ]
    )


===# ============================================================
# 15. APPLICATION STATUS
# ============================================================

print("================================================")
print("       SYNDATA ARENA APPLICATION READY")
print("================================================")
print("CSV Dataset Upload: READY")
print("BLIP Local Image Upload: READY")
print("CSV Analysis: ORIGINAL FUNCTION PRESERVED")
print("Image-Specific Analysis: CONNECTED")
print("Four Visualizations: READY")
print("Synthetic Data Generation: READY")
print("Quality Evaluation: READY")
print("ML Model Testing: READY")
print("Classification and Regression: SUPPORTED")
print("SynBot AI: READY")
print("Complete PDF Report: READY")
print("PDF ML Re-Run: DISABLED")
print("================================================")
print("Total Tabs: 7")
print("Gemini Dependency: DISABLED")
print("Public URL: ENABLED")
print("================================================")


# ============================================================
# 16. LAUNCH PUBLIC APPLICATION
# =========================================================

print("\nLaunching SynData Arena...")
print("Please wait for your public Gradio URL.")

demo_final.queue().launch(
    share=True,
    debug=True
)
```
